1. Créez un programme qui gère les commandes de café en fonction des différentes options telles que la taille, le type de
café, les extras, etc. Utilisez des correspondances pour traiter chaque option et calculer le prix total de la commande.


```python
print("--- Commande de Café ---")
taille = input("Taille (S, M, L) : ").upper()
type_cafe = input("Type (Noir, Latte, Cappuccino) : ").capitalize()

prix = 0

# Calcul du prix de base selon la taille
match taille:
    case "S": prix = 2.0
    case "M": prix = 3.0
    case "L": prix = 4.0
    case _: print("Taille inconnue, prix par défaut (M)"); prix = 3.0

# Supplément selon le type
match type_cafe:
    case "Latte": prix += 1.0
    case "Cappuccino": prix += 1.5
    case "Noir": prix += 0.0
    case _: print("Type inconnu, pas de supplément.")

print(f"Total à payer : {prix}€")
```

2. Créez un programme qui convertit une note numérique en une note alphabétique en utilisant une échelle de notation
standard. Utilisez des correspondances pour déterminer la note alphabétique correspondante en fonction de la note
numérique.

```python
note = float(input("Entrez la note (0-100) : "))

match note:
    case n if n >= 90: lettre = "A"
    case n if n >= 80: lettre = "B"
    case n if n >= 70: lettre = "C"
    case n if n >= 60: lettre = "D"
    case _: lettre = "F"

print(f"Votre note est : {lettre}")
```

3. Créez un programme qui génère un nombre aléatoire (import random) et permet à l'utilisateur de deviner ce nombre.
Utilisez des correspondances pour comparer la devinette de l'utilisateur avec le nombre généré et fournir des indices.


```python
import random

nombre_secret = random.randint(1, 20)
devinette = int(input("Devinez le nombre (1-20) : "))

match devinette:
    case _ if devinette == nombre_secret:
        print("Bravo ! Vous avez trouvé.")
    case _ if devinette > nombre_secret:
        print("Trop haut !")
    case _ if devinette < nombre_secret:
        print("Trop bas !")
```

4. Créez un programme qui calcule l'indice de masse corporelle (IMC) d'une personne en fonction de son poids et de sa
taille. Utilisez des correspondances pour interpréter et catégoriser l'IMC résultant en différentes catégories de poids.


```python
poids = float(input("Poids (kg) : "))
taille = float(input("Taille (m) : "))
imc = round(poids / (taille ** 2), 1)

match imc:
    case i if i < 18.5: cat = "Insuffisance pondérale"
    case i if i < 25: cat = "Poids normal"
    case i if i < 30: cat = "Surpoids"
    case _: cat = "Obésité"

print(f"Votre IMC est de {imc} ({cat}).")
```

6. Créez un programme qui génère des citations aléatoires à partir d'un nombre aléatoire. L'utilisateur devrait pouvoir
choisir un thème et le programme générera une citation aléatoire correspondante. Utilisez des correspondances pour
gérer les différents thèmes et générer les citations appropriées.

```python
import random

theme = input("Choisissez un thème (Sport, Motivation, Code) : ").capitalize()
index = random.randint(1, 2) # On a 2 citations par thème

match theme:
    case "Sport":
        citation = "Le sport va chercher la peur pour la dominer."
        if index == 1:
            citation = "Plus vite, plus haut, plus fort."
    case "Motivation":
        citation = "N'arrêtez jamais d'apprendre." if index == 1 else "Le succès est un voyage, pas une destination."
    case "Code":
        citation = "Ça marche sur ma machine." if index == 1 else "Le code ne ment jamais, les commentaires si."
    case _:
        citation = "Thème inconnu."

print(f"Citation ({theme}) : {citation}")
```
