Écrivez un programme Python
convertissant un nombre de secondes en jours, heures, minutes et secondes correspondants.

Exemple : 4561 secondes correspondent à 0 jour, 1 heure, 16 minutes et 1 seconde.
Réfléchissez à la méthode à utiliser,
puis réalisez l'algorithme en Python.
Testez-le ensuite pour vérifier sa validité.

```python
input_secondes = int(input("Entrez un nombre de secondes : ")) # Exemple: 90061

# 1. Calcul des jours (86400 secondes par jour)
days = input_secondes // 86400 # days = 1
reste = input_secondes % 86400 # reste = 3661

# 2. Calcul des heures dans le reste (3600 secondes par heure)
hours = reste // 3600 # 1
reste = reste % 3600 # 61

# 3. Calcul des minutes dans le reste restant (60 secondes par minute)
minutes = reste // 60 # 1
secondes = reste % 60 # 1

# Affichage avec f-string (plus simple pour mélanger texte et variables)
print(f"{input_secondes} secondes donnent :")
print(f"{days} jours, {hours} heures, {minutes} minutes et {secondes} secondes.")
# 1 jour, 1 heure, 1 minute et 1 seconde.
```
