1. Commencez par créer une liste de 10 nombres aléatoires compris entre 1 et 100. Ensuite, affichez cette liste générée.
Après cela, calculez la somme de tous les éléments de la liste et affichez-la.


```python
import random

# 10 nombres aléatoires entre 1 et 100 (randint : bornes incluses)
nombres = []
for i in range(10):
    nombres.append(random.randint(1, 100))

print(f"Liste générée : {nombres}")

# Somme de tous les éléments : on accumule dans un total
total = 0
for nombre in nombres:
    total += nombre

print(f"Somme des éléments : {total}")
# (la fonction intégrée sum(nombres) donne le même résultat en une ligne)
```

3. Générez deux ensembles de nombres aléatoires compris entre 1 et 20. Affichez ces deux ensembles générés. Enfin,
trouvez l'intersection des deux ensembles et affichez-la.

```python
import random

# Deux ensembles de nombres aléatoires entre 1 et 20.
# On fait 10 tirages chacun : le set élimine les doublons,
# il peut donc contenir MOINS de 10 éléments !
ensemble_1 = set()
ensemble_2 = set()

for i in range(10):
    ensemble_1.add(random.randint(1, 20))
    ensemble_2.add(random.randint(1, 20))

print(f"Ensemble 1 : {ensemble_1}")
print(f"Ensemble 2 : {ensemble_2}")

# L'intersection : les nombres présents dans LES DEUX ensembles
print(f"Intersection : {ensemble_1 & ensemble_2}")
```

4. Créez un dictionnaire contenant les prix de quelques fruits tels que la pomme, la banane et l'orange. Demandez à
l'utilisateur de saisir le nom d'un fruit, puis affichez le prix correspondant à ce fruit s'il existe dans le dictionnaire.

```python
prix_fruits = {
    "pomme": 0.75,
    "banane": 1.20,
    "orange": 0.90,
}

fruit = input("Entrez le nom d'un fruit : ").lower()

# On vérifie que la clé existe AVANT de lire (sinon : KeyError !)
if fruit in prix_fruits:
    print(f"Le prix de la {fruit} est {prix_fruits[fruit]} €")
else:
    print(f"{fruit} n'est pas dans le dictionnaire")
```

5. Créez une liste de tuples contenant le nom et l'âge de trois personnes. Trouvez ensuite la personne la plus âgée et
affichez son nom.


```python
personnes = [
    ("Alice", 30),
    ("Bob", 45),
    ("Chloé", 27),
]

# On part de la première personne, puis chaque personne
# plus âgée rencontrée prend sa place
nom_plus_age, age_plus_age = personnes[0]

for nom, age in personnes:      # unpacking : chaque tuple se déballe en nom, age
    if age > age_plus_age:
        nom_plus_age = nom
        age_plus_age = age

print(f"La personne la plus âgée est {nom_plus_age} ({age_plus_age} ans)")
```

6. Générez une liste de 10 nombres aléatoires compris entre 1 et 50. Affichez cette liste générée. Ensuite, filtrez les
nombres pairs de la liste et créez une nouvelle liste ne contenant que ces nombres pairs. Enfin, affichez la nouvelle
liste contenant uniquement les nombres pairs

```python
import random

nombres = []
for i in range(10):
    nombres.append(random.randint(1, 50))

print(f"Liste générée : {nombres}")

# Filtrage : on ne garde que les pairs (reste de la division par 2 = 0)
pairs = []
for nombre in nombres:
    if nombre % 2 == 0:
        pairs.append(nombre)

print(f"Nombres pairs : {pairs}")
```

8. Créez un dictionnaire de listes représentant différents cours et les étudiants inscrits dans chaque cours. Ajoutez des
étudiants à chaque cours. Ensuite, demandez à l'utilisateur de saisir le nom d'un cours et affichez la liste des
étudiants inscrits à ce cours.

9. Créez une liste de tuples représentant les commandes d'achat avec les produits et les quantités. Ensuite, créez un
dictionnaire de prix pour chaque produit. Calculez ensuite le coût total de toutes les commandes et affichez-le.

```python
commandes = [
    ("clavier", 2),
    ("souris", 3),
    ("écran", 1),
]

prix = {
    "clavier": 49.99,
    "souris": 19.99,
    "écran": 219.00,
}

# Pour chaque commande : quantité x prix du produit, accumulé dans le total
cout_total = 0
for produit, quantite in commandes:     # unpacking du tuple (produit, quantité)
    cout_total += quantite * prix[produit]

print(f"Coût total des commandes : {cout_total:.2f} €")
```

10. Créez une liste de dictionnaires représentant les informations des employés avec leur nom, salaire et département.
Calculez la somme des salaires pour chaque département, puis calculez la moyenne des salaires pour chaque
département. Enfin, affichez les moyennes des salaires pour chaque département.

```python
employes = [
    {"nom": "Alice", "salaire": 3200, "departement": "IT"},
    {"nom": "Bob", "salaire": 2800, "departement": "RH"},
    {"nom": "Chloé", "salaire": 3600, "departement": "IT"},
    {"nom": "David", "salaire": 3000, "departement": "RH"},
    {"nom": "Emma", "salaire": 4100, "departement": "IT"},
]

# Étape 1 : somme des salaires et nombre d'employés par département
sommes = {}
effectifs = {}

for employe in employes:
    departement = employe["departement"]

    if departement not in sommes:       # première fois qu'on voit ce département
        sommes[departement] = 0
        effectifs[departement] = 0

    sommes[departement] += employe["salaire"]
    effectifs[departement] += 1

print(f"Sommes par département : {sommes}")

# Étape 2 : moyenne = somme / nombre d'employés du département
for departement in sommes:
    moyenne = sommes[departement] / effectifs[departement]
    print(f"Salaire moyen du département {departement} : {moyenne:.2f} €")
```
