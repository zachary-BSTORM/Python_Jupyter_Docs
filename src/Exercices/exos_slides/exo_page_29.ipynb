4. Demandez à l'utilisateur deux nombres et affichez leur somme avec une fin de ligne personnalisée.

```python
nb1 = int(input("Entrez un premier nombre :"))
nb2 = int(input("Entrez un deuxieme nombre :"))

result = nb1 + nb2

print(result, end="!")
print(type(result))
```

```python
nom = input("Entrez votre nom :")
prenom = input("Entrez votre prenom :")
age = input("Entrez votre age :")

print("Vous êtes " + nom + " - " + prenom + ", Vous avez " + age + " ans")
```
