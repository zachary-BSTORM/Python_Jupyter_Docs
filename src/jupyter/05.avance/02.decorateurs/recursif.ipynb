# Récursivité 

- La récursivité consiste à implémenter une methode qui va pouvoir s'appeller elle même

- Il faut faire attention au condition de sortie 

- et il faut faire attention au limite de récursion qui renvoie une erreur 

---

```python
# exemple de récursif : fibo(40) prend 17s à s'exécuter
def fibo(n):
    if n < 2:
        return n
    return fibo(n - 1) + fibo(n - 2)

fibo(40)
```

```python
# exemple de récursif : fibo(40) avec cache s'exécute immédiatement
from functools import cache

@cache
def fibo_cache(n):
    if n < 2:
        return n
    return fibo_cache(n - 1) + fibo_cache(n - 2)

fibo_cache(40)
```

---

### Avantage et inconvenient avec la récursivité

- Les avantages : 
    - Permet de raccourcir le code 
    - Simplifie la manipulation en profondeur
    - Peut devenir indispensable dans certains cas ( backtracking )

- inconvénients :
    - On retrouve une limite de récursion (+- 1000 appels)
    - peut devenir plus gourmand en mémoire
    - Peut nécéssité d'utiliser d'autre concepts ( memoisation)
    - peut devenir plus difficile à déboguer

---

### Exemple Somme Récursive

```python
liste = [1, 2, 3, 4, 5]

def somme_recursif(liste):
    if len(liste) == 0:
        return 0
    return liste[0] + somme_recursif(liste[1:])

somme_recursif(liste)
```
