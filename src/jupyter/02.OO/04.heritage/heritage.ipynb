# L'héritage

L'héritage est un concept fondamental de la POO qui permet à une classe (**enfant**) de **réutiliser** les attributs et méthodes d'une autre classe (**parent**).

### Avantages
- **Réutilisabilité** : évite la duplication de code
- **Extensibilité** : la sous-classe peut ajouter ses propres fonctionnalités
- **Modularité** : facilite l'organisation et la maintenance du code

---

## Mise en place

1. **Définir la classe parente** avec les attributs et méthodes communs
2. **Déclarer la classe enfant** en indiquant la classe parente entre parenthèses
3. **Personnaliser** la sous-classe en ajoutant de nouveaux attributs ou méthodes

```python
class Vehicule:
    def __init__(self, marque, modele):
        # Attributs d'instance : chaque objet a ses propres valeurs
        self.marque = marque
        self.modele = modele

    def afficher_info(self):
        print(f"Marque: {self.marque}")
        print(f"Modèle: {self.modele}")


class Voiture(Vehicule):  # Voiture hérite de Vehicule
    pass                  # rien à ajouter : elle hérite du constructeur et des méthodes


# Voiture n'a rien défini elle-même : le constructeur (__init__) et la méthode
# afficher_info viennent entièrement de la classe parente Vehicule
ma_voiture = Voiture("Toyota", "Corolla")

print("Informations de la voiture :")
ma_voiture.afficher_info()  # Méthode héritée de Vehicule
```

---

## La méthode `super()`

`super()` permet d'appeler une méthode de la **classe parente** depuis la sous-classe. C'est particulièrement utile pour :
- Appeler le constructeur parent
- Étendre une méthode héritée sans la remplacer entièrement

```python
class Voiture(Vehicule):
    def __init__(self, marque, modele, couleur):
        super().__init__(marque, modele)   # appelle le constructeur de Vehicule
        self.couleur = couleur             # puis ajoute son propre attribut

    def afficher_info(self):
        super().afficher_info()            # étend la méthode héritée sans la remplacer
        print(f"Couleur: {self.couleur}")


# Instanciation : on observe le constructeur parent + l'extension de la méthode
ma_voiture = Voiture("Toyota", "Corolla", "Rouge")
ma_voiture.afficher_info()
```

## Héritage multiple

Contrairement à d'autres langages (C#, Java), Python permet l'**héritage multiple** : une classe peut hériter de **plusieurs** classes parentes.

> ⚠️ En cas de conflit de méthodes, Python utilise le **MRO** (Method Resolution Order) pour déterminer l'ordre de priorité.

```python
class Vehicule:
    def __init__(self, marque):
        self.marque = marque

    def afficher_marque(self):
        print(f"Marque du vehicule : {self.marque}")


class Modele:
    def afficher_modele(self):
        # Lit l'attribut d'instance `modele`, défini par la classe qui hérite
        print(f"Modele du vehicule : {self.modele}")


class Voiture(Vehicule, Modele):  # héritage multiple
    def __init__(self, marque, modele):
        super().__init__(marque)  # initialise la partie Vehicule
        self.modele = modele      # attribut utilisé par Modele.afficher_modele

    def afficher_info(self):
        self.afficher_marque()   # héritée de Vehicule
        self.afficher_modele()   # héritée de Modele


# Création d'une instance de la classe Voiture
ma_voiture = Voiture("Toyota", "Corolla")

# Affichage des informations de la voiture
ma_voiture.afficher_info()
```

# héritage multiple avec ** **kwargs**
- les parents prévoient un argument kwargs 
- l'enfant distribue les arguments aux parents
- avant de redescendre dans l'ordre hierarchique de l'héritage jusqu'a la classe concerné

```python
class Animal:
    def __init__(self,name : str,color : str , age : int,**kwargs):
        self.name = name
        self.color = color
        self.age = age
        
        
class Breed:
    def __init__(self,breed_name : str , price : int,**kwargs):
        self.breed_name = breed_name
        self.price = price
        
    
class Chien(Animal,Breed):
    def __init__(self,toy : str,hungry : bool,**kwargs):
        self.toy = toy
        self.hungry = hungry
        Breed.__init__(self,**kwargs)
        Animal.__init__(self,**kwargs)

    
rex = Chien("bones",True,name="rex",color="brun",age=5,breed_name="berger allemand",price=500)
```

## Redéfinition de méthodes (Override)

La redéfinition permet de **remplacer** l'implémentation d'une méthode héritée dans la classe enfant, sans modifier la classe parente.

Cela permet de personnaliser le comportement pour chaque sous-classe.

```python
class Vehicule:
    def demarrer(self):
        print("Le véhicule démarre.")
        
    def arreter(self):
        print("Le véhicule s'arrête.")


class Voiture(Vehicule):
    def demarrer(self):  # Redéfinition de la méthode
        print("La voiture démarre en appuyant sur l'accélérateur.")


ma_voiture = Voiture()
ma_voiture.demarrer()  # Appelle la version redéfinie
ma_voiture.arreter()   # Appelle la version héritée
```

## Astuce : `issubclass()`

La fonction `issubclass(enfant, parent)` vérifie si une classe est une **sous-classe** d'une autre. Renvoie `True` ou `False`.

```python
class Vehicule:
    pass

class Voiture(Vehicule):
    pass

class Moto(Vehicule):
    pass


# Verification si voiture est une sous-classe de vehicule
print(issubclass(Voiture, Vehicule))  # Affiche True

# Verification si moto est une sous-classe de vehicule
print(issubclass(Moto, Vehicule))  # Affiche True

# Verification si vehicule est une sous-classe de Moto
print(issubclass(Vehicule, Moto))  # Affiche False
```

## MRO — Method Resolution Order (C3 Linearization)

En héritage multiple, Python utilise l'algorithme **C3 Linearization** pour déterminer l'ordre dans lequel les classes sont parcourues lors de la recherche d'une méthode.

L'ordre est : **classe elle-même → parents de gauche à droite → ancêtres communs**

On peut consulter le MRO d'une classe avec `NomClasse.__mro__` ou `NomClasse.mro()`.

```python
class A:
    def methode(self):
        print("A.methode")

class B(A):
    def methode(self):
        print("B.methode")

class C(A):
    def methode(self):
        print("C.methode")

class D(B, C):   # héritage multiple
    pass


d = D()
d.methode()          # B.methode — B est avant C dans la définition de D

# Consulter le MRO
print(D.__mro__)
# (<class 'D'>, <class 'B'>, <class 'C'>, <class 'A'>, <class 'object'>)

print([cls.__name__ for cls in D.mro()])
# ['D', 'B', 'C', 'A', 'object']
```

## `super()` — appel coopératif

`super()` retourne un **proxy** de la classe parente selon le MRO. En héritage multiple, `super()` est **coopératif** : chaque classe doit appeler `super()` pour que toutes les classes parentes soient correctement initialisées.

Signature :
- `super()` — forme courte (Python 3+), recommandée
- `super(ClasseActuelle, self)` — forme explicite (Python 2 compatible)

```python
class Base:
    def __init__(self):
        print("Base.__init__")
        super().__init__()   # appelle object.__init__

class MixinA:
    def __init__(self):
        print("MixinA.__init__")
        super().__init__()

class MixinB:
    def __init__(self):
        print("MixinB.__init__")
        super().__init__()

class Enfant(MixinA, MixinB, Base):
    def __init__(self):
        print("Enfant.__init__")
        super().__init__()

e = Enfant()
# Affiche dans l'ordre MRO :
# Enfant.__init__
# MixinA.__init__
# MixinB.__init__
# Base.__init__

print(Enfant.__mro__)
# Enfant → MixinA → MixinB → Base → object
```
