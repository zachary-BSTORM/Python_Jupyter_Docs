# Les classes et les objets

Pour créer une classe en Python, on utilise le mot-clé `class` suivi du nom de la classe.

### Conventions de nommage

| Élément | Convention | Exemple |
|---|---|---|
| Classe | **PascalCase** | `Animal`, `VoitureSport` |
| Variable / Fonction | **snake_case** | `ma_variable`, `calculer_somme` |
| Fichier | **snake_case** | `les_class.py` |

### Concepts clés

- Une **classe** est un **plan de construction** pour créer des objets
- Un **objet** est une **instance** d'une classe
- Une classe contient des **attributs** (variables) et des **méthodes** (fonctions)
- Les **attributs** décrivent les caractéristiques de l'objet

**Exemples :**
- Classe `Voiture` → attributs : `couleur`, `marque`, `modele`
- Classe `Utilisateur` → attributs : `nom`, `age`, `email`
- Classe `Animal` → attributs : `nom`, `couleur`, `age`

---

## Création d'une classe vide

Le mot-clé `pass` permet de définir une classe (ou fonction) vide, sans contenu :

```python
class Animal():
    pass
```

> `pass` est un mot-clé qui ne fait rien. Il sert de **placeholder** quand un bloc de code est requis mais qu'on n'a pas encore d'implémentation.

---

## Le constructeur `__init__`

Le constructeur est une méthode spéciale appelée **automatiquement** lors de la création d'un objet.

- Le mot-clé **`self`** fait référence à l'instance en cours de création
- On initialise les **attributs d'instance** dans le constructeur
- Chaque instance possède ses propres valeurs d'attributs

```python
class Animal():
    def __init__(self, nom, couleur, age):
        # Variables d'instance : chaque objet a ses propres valeurs
        self.nom = nom
        self.couleur = couleur
        self.age = age
```

---

## Les méthodes

Les méthodes sont des fonctions définies à l'intérieur d'une classe. Elles reçoivent toujours `self` comme premier paramètre :

```python
class Animal():
    def __init__(self, nom, couleur, age):
        self.nom = nom
        self.couleur = couleur
        self.age = age
    
    def communiquer(self):
        print(f"Je suis un animal, je m'appelle {self.nom}, je suis {self.couleur} et j'ai {self.age} ans")
```

## Exemple complet : la classe `Chien`

La classe ci-dessous regroupe tout ce qu'on vient de voir : une docstring, un constructeur et des méthodes.

> **Remarque importante** : les attributs `nom`, `age` et `couleur` sont définis **uniquement dans `__init__`**, pas dans le corps de la classe. Ce n'est pas équivalent : un attribut défini dans le corps de la classe est un **attribut de classe**, partagé par toutes les instances, alors qu'un attribut défini dans `__init__` appartient à **chaque objet** individuellement (voir la section « Attributs de classe vs attributs d'instance » plus bas).

Les méthodes sont nommées en **snake_case** (`communiquer`, `manger`, `dormir`), conformément aux conventions Python vues en début de chapitre.

```python
class Chien():
    '''
    Classe qui représente un chien.
    '''

    # Constructeur : méthode spéciale appelée automatiquement
    # lors de la création d'un objet de la classe.
    # Les attributs d'instance sont définis ici, et nulle part ailleurs :
    # chaque objet possède ainsi ses propres valeurs.
    def __init__(self, nom, age, couleur):
        self.nom = nom
        self.age = age
        self.couleur = couleur

    # Méthodes : des fonctions qui appartiennent à la classe (en snake_case)
    def communiquer(self):
        print(f"Je suis un chien je m'appelle {self.nom} et je suis de couleur {self.couleur} et j'ai {self.age} ans")

    def manger(self):
        print(f"{self.nom} mange des croquettes")

    def dormir(self):
        print(f"{self.nom} dort sur le canapé vive les poils")
```

## Instanciation

**Instancier** une classe, c'est créer un objet concret à partir du plan de construction. On appelle la classe comme une fonction, en lui passant les arguments attendus par `__init__`.

> Si la classe était définie dans un autre fichier (par exemple `les_class.py`), on l'importerait avec `from les_class import Animal`. Ici, les classes sont définies directement dans le notebook : aucun import n'est nécessaire.

```python
# Instanciation de la classe Animal définie plus haut
milo = Animal("Milo", "gris", 3)

# On accède aux attributs et aux méthodes de l'objet avec le point
print(milo.nom)
milo.communiquer()
```

---

## Le `if __name__ == "__main__"`

Cette condition vérifie que le fichier est exécuté **directement** (et non importé comme module). C'est une bonne pratique pour séparer la logique de test du code réutilisable :

```python
if __name__ == "__main__":
    # Instanciation de la classe Chien : le constructeur reçoit les arguments
    rex = Chien("Rex", 5, "roux")

    # On appelle les méthodes de l'objet à partir du moment
    # où nous avons une instance de la classe
    rex.communiquer()
    rex.manger()
    rex.dormir()
```

## Attributs de classe vs attributs d'instance

| | Attribut de **classe** | Attribut d'**instance** |
|---|---|---|
| Déclaré | Dans le corps de la classe (hors `__init__`) | Dans `__init__` (via `self.x = ...`) |
| Partagé | Oui — entre **toutes** les instances | Non — propre à chaque objet |
| Accès | `NomClasse.attr` ou `self.attr` | `self.attr` |
| Mutation | Modifier via `NomClasse.attr` pour toutes | Modifier via `self.attr` pour une seule |

> **Piège** : si on fait `self.attr = valeur` et qu'un attribut de classe du même nom existe, Python crée un **attribut d'instance** qui masque (shadow) l'attribut de classe. La classe n'est pas modifiée.

```python
class Voiture:
    # Attribut de classe (partagé)
    nombre_roues = 4
    compteur = 0

    def __init__(self, marque):
        # Attributs d'instance (propres à chaque objet)
        self.marque = marque
        Voiture.compteur += 1   # incrémente le compteur partagé

v1 = Voiture("Toyota")
v2 = Voiture("Honda")

print(Voiture.nombre_roues)  # 4 — accès via la classe
print(v1.nombre_roues)       # 4 — accès via une instance
print(Voiture.compteur)      # 2 — partagé entre toutes les instances

# Piège : masquage
v1.nombre_roues = 3          # crée un attribut d'instance sur v1 seulement
print(v1.nombre_roues)       # 3 — attribut d'instance de v1
print(v2.nombre_roues)       # 4 — attribut de classe inchangé
print(Voiture.nombre_roues)  # 4 — classe inchangée
```

## Introspection — inspecter les objets

Python offre plusieurs outils pour examiner les objets à l'exécution :

| Fonction | Description |
|---|---|
| `type(obj)` | Retourne la classe de l'objet |
| `isinstance(obj, class)` | Vérifie si `obj` est une instance de `class` (ou sous-classe) |
| `issubclass(cls1, cls2)` | Vérifie si `cls1` est une sous-classe de `cls2` |
| `hasattr(obj, 'nom')` | Vérifie si l'attribut `'nom'` existe |
| `getattr(obj, 'nom', default)` | Retourne l'attribut `'nom'` (ou `default` si absent) |
| `setattr(obj, 'nom', valeur)` | Définit l'attribut `'nom'` |
| `delattr(obj, 'nom')` | Supprime l'attribut `'nom'` |
| `dir(obj)` | Liste tous les attributs et méthodes |
| `vars(obj)` | Retourne `__dict__` de l'objet |
| `id(obj)` | Identifiant unique de l'objet (adresse mémoire) |

```python
class Animal:
    def __init__(self, nom):
        self.nom = nom

class Chien(Animal):
    pass

rex = Chien("Rex")

print(type(rex))                   # <class '__main__.Chien'>
print(isinstance(rex, Chien))      # True
print(isinstance(rex, Animal))     # True — Chien est sous-classe d'Animal
print(issubclass(Chien, Animal))   # True

print(hasattr(rex, 'nom'))         # True
print(hasattr(rex, 'age'))         # False

print(getattr(rex, 'nom'))         # 'Rex'
print(getattr(rex, 'age', 0))      # 0 (valeur par défaut)

setattr(rex, 'age', 5)             # équivalent à rex.age = 5
print(vars(rex))                   # {'nom': 'Rex', 'age': 5}

delattr(rex, 'age')                # équivalent à del rex.age
```
