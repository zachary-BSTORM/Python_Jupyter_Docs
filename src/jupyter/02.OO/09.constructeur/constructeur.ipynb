# Les constructeurs

Le constructeur est une méthode spéciale qui **initialise** les objets dès leur création. Il est essentiel pour :
- Configurer l'**état initial** des attributs
- Effectuer des **vérifications** à la création
- Préparer l'objet à être utilisé

## Syntaxe : `__init__`

En Python, le constructeur est la méthode `__init__`. Elle est appelée **automatiquement** à chaque instanciation.

Le premier paramètre est toujours `self` (référence à l'objet en cours de création).

```python
class Voiture:
    def __init__(self):
        pass
```

---

## Constructeur avec paramètres

Les paramètres passés lors de l'instanciation sont transmis au constructeur pour initialiser les attributs :

```python
class Voiture:
    def __init__(self,marque, modele, annee):
        self.marque = marque
        self.modele = modele
        self.annee = annee
        
voiture1 = Voiture("Toyota", "Corolla", 2020)
voiture2 = Voiture("Honda", "Civic", 2019)
voiture3 = Voiture("Ford", "Focus", 2021)
```

---

## Sans constructeur : l'héritage de `object`

Si une classe ne définit **aucun** `__init__`, Python n'en « génère » pas un : la classe **hérite** simplement du `__init__` de `object`, la classe mère de toutes les classes en Python. Ce constructeur hérité ne prend aucun argument et n'initialise aucun attribut. Dès que la classe définit son propre `__init__`, c'est lui qui est utilisé à la place.

> ⚠️ Python ne supporte **pas** la surcharge de constructeurs (plusieurs `__init__` dans la même classe). On utilise les **valeurs par défaut** pour simuler ce comportement.

### Exemple : constructeur sans paramètre

```python
class Exemple:
    def __init__(self):
        print("constructeur sans paramètre appelé")
```

### Exemple : constructeur avec paramètre

```python
class Exemple2:
    def __init__(self, nom):
        self.nom = nom
        print(f"constructeur avec paramètre appelé : {self.nom}")
```

```python
# Création avec le constructeur sans paramètre
exemple1 = Exemple()

# Création avec le constructeur avec paramètre
exemple2 = Exemple2("objet_A")

# Seul exemple2 possède un attribut 'nom' : Exemple n'en définit aucun
print("Nom de l'objet exemple2 :", exemple2.nom)
print("exemple1 a-t-il un attribut 'nom' ?", hasattr(exemple1, "nom"))
```

---

## Appel du constructeur parent avec `super()`

Lors de l'héritage, on appelle le constructeur de la classe parente avec `super().__init__()` pour initialiser les attributs hérités :

```python
class Vehicule:
    def __init__(self, marque):
        self.marque = marque

class Voiture(Vehicule):
    def __init__(self, marque, modele):
        super().__init__(marque)
        self.modele = modele
        print("Constructeur de la classe Voiture appelé")
        
voiture = Voiture("Toyota", "Corolla")

print("Marque de la voiture :", voiture.marque)
print("Modèle de la voiture :", voiture.modele)
```

---

## Astuce : valeurs par défaut

On peut définir des **valeurs par défaut** pour certains paramètres, permettant une instanciation flexible :

```python
class Voiture:
    def __init__(self, marque, modele = "Modèle par défaut"):
        self.marque = marque
        self.modele = modele
        print("Constructeur avec marque et modèle appelé")

voiture1 = Voiture("Toyota", "Corolla")
voiture2 = Voiture("Honda")

print("Marque de la voiture 1 :", voiture1.marque)
print("Modèle de la voiture 1 :", voiture1.modele)

print("Marque de la voiture 2 :", voiture2.marque)
print("Modèle de la voiture 2 :", voiture2.modele)
```

## Piège : valeur par défaut mutable

Une valeur par défaut est évaluée **une seule fois**, au moment où Python lit le `def` — pas à chaque appel. Avec une valeur **mutable** (modifiable : liste, dictionnaire, set...), c'est donc **le même objet** qui est partagé par tous les appels du constructeur : toutes les instances écrivent dans la même liste !

**Le remède** : utiliser `None` comme valeur par défaut, puis créer une nouvelle liste dans le constructeur.

```python
# ❌ Le piège : la liste par défaut est PARTAGÉE entre tous les objets
class PanierBogue:
    def __init__(self, items=[]):   # créée UNE seule fois, au def
        self.items = items

    def ajouter(self, article):
        self.items.append(article)


p1 = PanierBogue()
p2 = PanierBogue()

p1.ajouter("pomme")
print("Panier 1 :", p1.items)   # ['pomme']
print("Panier 2 :", p2.items)   # ['pomme'] aussi ! p2 partage la même liste


# ✅ Le remède : None par défaut, puis une liste neuve dans le constructeur
class Panier:
    def __init__(self, items=None):
        self.items = items if items is not None else []

    def ajouter(self, article):
        self.items.append(article)


p1 = Panier()
p2 = Panier()

p1.ajouter("pomme")
print("Panier 1 :", p1.items)   # ['pomme']
print("Panier 2 :", p2.items)   # [] — chaque objet a bien sa propre liste
```
