# Le polymorphisme

Le polymorphisme permet à des objets de **classes différentes** d'être traités de manière **uniforme** s'ils partagent une interface commune (mêmes noms de méthodes).

Concrètement, une même méthode peut avoir un **comportement différent** selon le type de l'objet sur lequel elle est appelée.

### Principe

On écrit du code qui manipule des objets via leur **interface commune**, sans se soucier de leur type exact. C'est Python qui appelle la bonne implémentation au moment de l'exécution.

```python
class Voiture:
    def description(self):
        return "Je suis une voiture standard."
    
class VoitureSport(Voiture):
    def description(self):
        return "Je suis une voiture de sport rapide."
    
class VoitureElectrique(Voiture):
    def description(self):
        return "Je suis une voiture électrique économe en carburant."
    
# Fonstion générique prenant une voiture en argument

def afficher_description(voiture):
    print(voiture.description())
    
# Création d'instances de chaque type de voiture
voiture_standard = Voiture()
voiture_sport = VoitureSport()
voiture_electrique = VoitureElectrique()

# Appel de la fonction avec chaque type de voiture

afficher_description(voiture_standard)  # Affiche "Je suis une voiture standard."
afficher_description(voiture_sport)     # Affiche "Je suis une voiture de sport rapide."
afficher_description(voiture_electrique)  # Affiche "Je suis une voiture électrique économe en carburant."
```

## Duck Typing et polymorphisme sans héritage

> *"Si ça marche comme un canard et que ça cancane comme un canard, alors c'est un canard."*

Python utilise le **duck typing** : l'appartenance à une interface est déterminée par les **méthodes disponibles**, pas par l'héritage. Deux classes qui n'ont aucun lien hiérarchique peuvent être utilisées de manière polymorphe si elles partagent les mêmes noms de méthodes.

```python
# Polymorphisme sans héritage — duck typing
class Chien:
    def parler(self):
        return "Woof!"

class Chat:
    def parler(self):
        return "Miaou!"

class Robot:
    def parler(self):
        return "Bzzzt!"

# Aucun héritage commun, mais tous ont .parler()
def faire_parler(entite):
    print(entite.parler())  # Python ne vérifie pas le type

for entite in [Chien(), Chat(), Robot()]:
    faire_parler(entite)    # fonctionne parfaitement


# Polymorphisme avec les types built-in
for conteneur in [[1, 2, 3], (4, 5, 6), {7, 8, 9}, "abc"]:
    print(f"{type(conteneur).__name__} : len={len(conteneur)}")
# list, tuple, set, str ont tous __len__ → polymorphisme via le data model
```

## Override complet vs extension avec `super()`

| Approche | Quand l'utiliser | Code |
|---|---|---|
| **Override complet** | Comportement totalement différent | Redéfinir sans appeler `super()` |
| **Extension** | Ajouter au comportement parent | Appeler `super().methode()` puis ajouter |

```python
class Animal:
    def __init__(self, nom):
        self.nom = nom

    def decrire(self):
        return f"Je suis {self.nom}"

    def parler(self):
        return "..."


class Chien(Animal):
    # Extension : ajoute au comportement parent
    def decrire(self):
        base = super().decrire()        # réutilise la logique parente
        return f"{base}, un chien fidèle"

    # Override complet : remplace totalement
    def parler(self):
        return "Woof!"


class Chaton(Animal):
    def __init__(self, nom, couleur):
        super().__init__(nom)           # initialise nom via le parent
        self.couleur = couleur          # ajoute l'attribut propre

    def decrire(self):
        return f"{super().decrire()}, un chaton {self.couleur}"

    def parler(self):
        return "Miaou!"


animaux = [Chien("Rex"), Chaton("Mimi", "gris"), Animal("??")]

for a in animaux:
    print(f"{a.decrire()} → {a.parler()}")
```
