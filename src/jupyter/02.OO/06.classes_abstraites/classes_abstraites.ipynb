# Les classes abstraites

Une classe abstraite sert de **modèle** pour d'autres classes. Elle définit un **contrat** que les sous-classes doivent respecter.

### Caractéristiques
- **Ne peut pas être instanciée** directement
- Peut contenir des **méthodes abstraites** (déclarées mais non implémentées)
- Peut contenir des **méthodes concrètes** (avec implémentation)
- Les sous-classes **doivent** implémenter toutes les méthodes abstraites

---

## Le module `abc`

Le module `abc` (*Abstract Base Classes*) est intégré à Python et fournit les outils nécessaires pour définir des classes abstraites.

- **`ABC`** : classe de base à hériter pour créer une classe abstraite
- **`@abstractmethod`** : décorateur pour marquer une méthode comme abstraite

```python
from abc import ABC, abstractmethod

class Vehicule(ABC):
    @abstractmethod
    def mouvement(self):
        pass
```

---

### La preuve : une classe abstraite ne s'instancie pas

C'est le point central du chapitre, vérifions-le : tenter de créer un objet `Vehicule` lève une **`TypeError`**. Et une sous-classe qui **n'implémente pas** toutes les méthodes abstraites reste abstraite elle aussi — elle est refusée de la même façon.

```python
# 1) Instancier directement la classe abstraite → refusé
try:
    v = Vehicule()
except TypeError as e:
    print("TypeError :", e)


# 2) Une sous-classe incomplète (mouvement non implémentée) reste abstraite → refusée aussi
class Trottinette(Vehicule):
    pass  # n'implémente pas mouvement()

try:
    t = Trottinette()
except TypeError as e:
    print("TypeError :", e)


# 3) Une sous-classe complète, elle, s'instancie sans problème
class Velo(Vehicule):
    def mouvement(self):
        return "Le vélo pédale"

v = Velo()
print(v.mouvement())
```

---

## Les décorateurs `@abstractmethod`

Le décorateur `@abstractmethod` marque une méthode comme **obligatoire à implémenter** dans les sous-classes. Sans implémentation, la sous-classe ne pourra pas être instanciée.

```python
from abc import ABC, abstractmethod

class Vehicule(ABC):
    @abstractmethod
    def vitesse_max(self):
        pass

    @abstractmethod
    def mouvement(self):
        pass
```

---

## Implémentation complète

Chaque sous-classe **doit** fournir une implémentation de toutes les méthodes et propriétés abstraites pour pouvoir être instanciée :

> ⚠️ Pour déclarer une **propriété abstraite**, l'ordre des décorateurs est obligatoire : `@property` **au-dessus** de `@abstractmethod` (inverser les deux ne fonctionne pas).

```python
from abc import ABC, abstractmethod

class Vehicule(ABC):
    @property
    @abstractmethod
    def vitesse_max(self):
        pass

    @abstractmethod
    def mouvement(self):
        pass


class Voiture(Vehicule):
    _vitesse_actuelle = 0

    @property
    def vitesse_actuelle(self):
        return self._vitesse_actuelle

    @property
    def vitesse_max(self):
        return 200

    def mouvement(self):
        return "La voiture roule"


ma_voiture = Voiture()

print("Vitesse actuelle :", ma_voiture.vitesse_actuelle)
print("Vitesse maximale :", ma_voiture.vitesse_max)
print(ma_voiture.mouvement())
```

---

## Synthèse

| À retenir | En une phrase |
|---|---|
| Classe abstraite | Un **modèle** qui définit un contrat, jamais instancié directement |
| `ABC` | La classe de base du module `abc` dont on hérite pour rendre une classe abstraite |
| `@abstractmethod` | Marque une méthode **obligatoire** : sans elle, pas d'instanciation possible |
| Sous-classe incomplète | Reste abstraite : `TypeError` à l'instanciation tant que tout le contrat n'est pas implémenté |
| Propriété abstraite | `@property` puis `@abstractmethod` (dans cet ordre) |

> **En entretien** : « À quoi sert une classe abstraite ? » — À imposer un contrat commun à une famille de classes : elle garantit que chaque sous-classe implémente les méthodes essentielles, sinon Python refuse de créer l'objet (`TypeError`). C'est l'équivalent Python d'une classe abstraite Java/C#, fourni par le module `abc`.
