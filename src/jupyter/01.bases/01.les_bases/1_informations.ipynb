# Bases de l'informatique

Avant de plonger dans la programmation, il est utile de comprendre les bases du fonctionnement d'un ordinateur et les systèmes de numération utilisés en informatique.

---

## Les systèmes de numération

| Système | Base | Symboles | Exemple |
|---|---|---|---|
| **Binaire** | 2 | `0`, `1` | `1010` = 10 en décimal |
| **Octale** | 8 | `0` à `7` | `12` = 10 en décimal |
| **Décimal** | 10 | `0` à `9` | `42` |
| **Hexadécimal** | 16 | `0` à `9`, `A` à `F` | `2A` = 42 en décimal |

> **ASCII** est un encodage qui associe un nombre (0–127) à chaque caractère (lettres, chiffres, symboles).

## Les unités de mesure

| Unité | Abréviation | Équivalence |
|---|---|---|
| **Bit** | b | 0 ou 1 (unité de base) |

## Déplacement de bits (Bit Shifting)

Le déplacement de bits est une opération très rapide qui permet de **multiplier** ou **diviser** un nombre par une puissance de 2.

- `<<` : décalage à gauche → multiplie par 2ⁿ
- `>>` : décalage à droite → divise par 2ⁿ

**Exemples :**
- `00000001` (1) `<< 2` = `00000100` (4) → 1 × 2² = 4
- `00000100` (4) `>> 2` = `00000001` (1) → 4 ÷ 2² = 1
- `00000001` (1) `<< 3` = `00001000` (8) → 1 × 2³ = 8

---

## Conversion décimal → binaire

Pour convertir un nombre décimal en binaire, on soustrait les puissances de 2 décroissantes jusqu'à atteindre 0.

**Exemple : 741 en binaire**

| Position | 9 | 8 | 7 | 6 | 5 | 4 | 3 | 2 | 1 | 0 |
|---|---|---|---|---|---|---|---|---|---|---|
| Puissance | 2⁹ | 2⁸ | 2⁷ | 2⁶ | 2⁵ | 2⁴ | 2³ | 2² | 2¹ | 2⁰ |
| Valeur | 512 | 256 | 128 | 64 | 32 | 16 | 8 | 4 | 2 | 1 |
| Bit | 1 | 0 | 1 | 1 | 1 | 0 | 0 | 1 | 0 | 1 |

741 = 512 + 128 + 64 + 32 + 4 + 1 = **`1011100101`** en binaire

> 💡 Vérification en Python : `bin(741)` renvoie `'0b1011100101'`.

## Pourquoi connaître tout cela ?

Parce que c'est le **fondement de l'informatique**. Cela permet de comprendre :

- Comment un ordinateur **traite les données**
- Comment la **mémoire** est organisée
- Comment un **disque dur** stocke les informations
- Comment un **réseau** transmet les données

---

## Les composants clés d'un ordinateur

### Le processeur (CPU)
Le cerveau de l'ordinateur. Il contient des millions de transistors qui effectuent des opérations logiques et arithmétiques. La vitesse de traitement des données en dépend directement.

### La mémoire vive (RAM)
Mémoire utilisée pour stocker les données et programmes **en cours d'exécution**. C'est une mémoire **volatile** : elle perd son contenu lorsque l'ordinateur est éteint.

### Le cache
Mémoire très rapide située proche du processeur, qui stocke les données les plus fréquemment utilisées. C'est aussi une mémoire **volatile**.

### La mémoire morte (ROM)
Mémoire qui stocke les données et programmes qui **ne changent pas** (ex : le BIOS). C'est une mémoire **non volatile** : elle conserve son contenu même hors tension.

### Le disque dur (HDD / SSD)
Périphérique de stockage **non volatile** qui conserve les fichiers, programmes et le système d'exploitation de manière permanente.

### Les bus
Ensemble de fils qui transportent les données entre le processeur, la mémoire et les périphériques. Ils assurent la communication entre les différents composants.
