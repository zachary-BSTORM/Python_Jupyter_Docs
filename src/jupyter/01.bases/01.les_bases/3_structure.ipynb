# Informations sur le language **PYTHON**

- Python est un langage de programmation puissant et facile à apprendre.
- Il dispose de structures de données de haut niveau et permet une approche simple mais efficace de la programmation orientée objet.
- Parce que son typage est dynamique et qu'il est interprété, Python est un langage idéal pour l'écriture de scripts et le développement rapide d'applications dans de nombreux domaines et sur la plupart des plateformes.
- L'interpréteur Python peut être facilement étendu par de nouvelles fonctions et types de données implémentés en C ou C++ (ou tout autre langage appelable depuis le C).
- Python est également adapté comme langage d'extension pour personnaliser des applications.

# L'indentation en **PYTHON**

- **Python utilise l'indentation au lieu des accolades**

### Détails techniques

- L'indentation est **obligatoire** et fait partie de la **syntaxe** du langage (contrairement à C, Java, JavaScript).
- La convention officielle (**PEP 8**) recommande **4 espaces** par niveau d'indentation (pas de tabulations).
- Un bloc de code est défini par un `:` suivi d'une indentation cohérente.
- Une mauvaise indentation provoque une `IndentationError` ou une `TabError` à l'exécution.

```python
# ✅ Correct — 4 espaces
if True:
    print("bloc correctement indenté")

# ❌ Incorrect — mélange espaces/tabulations → TabError
if True:
	print("tabulation")
    print("espace")   # IndentationError
```

> ℹ️ En interne, l'interpréteur CPython utilise le niveau d'indentation pour construire l'**AST** (Abstract Syntax Tree), l'arbre syntaxique abstrait du programme.

# Structure d'un fichier Python

On peut exécuter des commandes Python directement dans la console, mais cela ne permet pas de **réutiliser** le code facilement. Il est donc préférable de l'écrire dans un **fichier `.py`**.

Pour exécuter un fichier, on se place dans le répertoire correspondant et on lance :
```bash
python nom_du_fichier.py
```

## Les commentaires

Les commentaires permettent de documenter le code. Ils sont **ignorés** par l'interpréteur Python.

- **Commentaire sur une ligne** : avec le caractère `#`
- **Commentaire multiligne** : avec des triples guillemets `"""` ou `'''`

## Organisation d'un fichier

Pour garder un code propre et lisible, on organise généralement un fichier en **sections** :

1. **Les imports** — les modules et bibliothèques nécessaires
2. **Les variables** — les constantes et variables globales
3. **Les fonctions** — les blocs de logique réutilisables
4. **Le code principal** — le point d'entrée du programme

## 🔧 Technique — Les imports en détail

Python distingue **3 types d'imports** :

| Type | Syntaxe | Exemple |
|------|---------|--------|
| Module standard (stdlib) | `import module` | `import math` |
| Import partiel | `from module import nom` | `from math import sqrt` |
| Alias | `import module as alias` | `import numpy as np` |

### Ordre recommandé par PEP 8

```python
# 1. Bibliothèques standard Python
import os
import math

# 2. Bibliothèques tierces (installées via pip)
import numpy as np
import pandas as pd

# 3. Modules locaux (fichiers de votre propre projet)
from mon_module import ma_fonction
```

> ⚠️ **Éviter** `from module import *` — cela importe tout dans le namespace global et peut créer des conflits de noms difficiles à déboguer.

### Fonctionnement interne

Quand Python importe un module, il :
1. Cherche le fichier dans `sys.path` (liste de répertoires)
2. Compile le fichier en **bytecode** (fichier `.pyc` dans `__pycache__/`)
3. Exécute le module une seule fois et le met en cache dans `sys.modules`

```python
import sys
print(sys.path)       # Liste des chemins de recherche
print(sys.modules)    # Modules déjà chargés en mémoire
```

## 🔧 Technique — Docstrings & Annotations de types

### Les Docstrings

Une **docstring** est une chaîne de caractères placée en première instruction d'une fonction, classe ou module. Elle est accessible via l'attribut `__doc__`.

```python
def addition(args: list) -> float:
    """
    Calcule la somme de tous les éléments d'une liste.

    Args:
        args (list): Liste de nombres à additionner.

    Returns:
        float: La somme des éléments.

    Example:
        >>> addition([1, 2, 3])
        6
    """
    return sum(args)

print(addition.__doc__)   # Affiche la docstring
help(addition)            # Affiche la docstring formatée
```

### Les Annotations de types (Type Hints — PEP 484)

Depuis Python 3.5, on peut annoter les types des paramètres et de la valeur de retour. Ces annotations sont **facultatives** et non vérifiées à l'exécution, mais utilisées par les linters et IDE.

```python
# Sans annotation
def addition(args):
    ...

# Avec annotation
def addition(args: list[int | float]) -> float:
    ...

# Variable annotée
total: int = 0
nom: str = "Python"
```

| Outil | Rôle |
|-------|------|
| `mypy` | Vérificateur de types statique |
| `pyright` | Vérificateur rapide (utilisé par VS Code) |
| `typing` | Module standard pour les types avancés (`List`, `Dict`, `Optional`...) |

## 🔧 Technique — Le bloc `if __name__ == "__main__"`

C'est une convention **fondamentale** en Python pour séparer le code exécutable du code importable.

### Comment ça fonctionne ?

Quand Python exécute un fichier :
- **Directement** (`python fichier.py`) → `__name__` vaut `"__main__"`
- **Via un import** (`import fichier`) → `__name__` vaut `"fichier"`

```python
# ===================================================
# Les imports
# ===================================================
import math

# ===================================================
# Les fonctions
# ===================================================
def addition(args: list) -> float:
    """Retourne la somme des éléments d'une liste."""
    return sum(args)

# ===================================================
# Point d'entrée principal
# ===================================================
if __name__ == "__main__":
    resultat = addition([1, 2, 3])
    print(f"Résultat : {resultat}")
```

> ✅ **Bonne pratique** : toujours encapsuler le code principal dans `if __name__ == "__main__":` pour que le fichier puisse être **importé** sans effets de bord.

### Sans ce bloc — problème classique

```python
# module_a.py
print("Ce message s'affiche même si on fait juste 'import module_a' !")  # ❌
```

## 🔧 Technique — Conventions PEP 8 (style de code)

**PEP 8** est le guide de style officiel de Python. Il définit des règles pour rendre le code lisible et cohérent.

### Nommage

| Élément | Convention | Exemple |
|---------|------------|--------|
| Variable / Fonction | `snake_case` | `ma_variable`, `calculer_total()` |
| Constante | `UPPER_CASE` | `MAX_TAILLE = 100` |
| Classe | `PascalCase` | `MaClasse`, `CalculateurImpots` |
| Module / Fichier | `snake_case` | `mon_module.py` |
| Attribut privé | `_underscore` | `_valeur_interne` |

### Longueur de ligne

- Maximum **79 caractères** par ligne (ou 99 selon les projets)
- Utiliser `\` ou des parenthèses pour continuer sur la ligne suivante

```python
# Retour à la ligne avec parenthèses
resultat = (valeur_un
            + valeur_deux
            + valeur_trois)
```

### Espacement

```python
# ✅ Correct
x = 1
y = x + 2
liste = [1, 2, 3]

# ❌ Incorrect
x=1
y=x+2
liste=[1,2,3]
```

| Outil | Rôle |
|-------|------|
| `flake8` | Vérifie le respect de PEP 8 |
| `black` | Formate automatiquement le code |

pip install flake8 black isort

````python
"""Exemple de bonne structure de fichier Python : calculs simples sur une liste de notes."""

# ===================================================
# Les imports
# ===================================================
from statistics import mean

# ===================================================
# Les constantes
# ===================================================
NOTE_MAXIMALE = 20

# ===================================================
# Les fonctions
# ===================================================
def addition(nombres: list[float]) -> float:
    """Retourne la somme des éléments de la liste."""
    return sum(nombres)


def moyenne(nombres: list[float]) -> float:
    """Retourne la moyenne des éléments de la liste."""
    return mean(nombres)


# ===================================================
# Le code principal
# ===================================================
if __name__ == "__main__":
    notes = [12, 15.5, 9, 18]

    print(f"Notes      : {notes}")
    print(f"Total      : {addition(notes)}")
    print(f"Moyenne    : {moyenne(notes)} / {NOTE_MAXIMALE}")
````
