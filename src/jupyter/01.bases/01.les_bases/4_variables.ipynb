# Déclaration de variables

Les variables sont des **espaces de stockage** qui permettent de conserver des valeurs en mémoire. Le **type** détermine l'espace de stockage utilisé.

### Règles de nommage
- Les noms doivent commencer par une **lettre** ou un **underscore** (`_`)
- Ils peuvent contenir que des lettres, des chiffres et des underscores
- Ils **ne peuvent pas** commencer par un chiffre
- La convention Python est le **`snake_case`** (ex : `ma_variable`, `nombre_total`)
- Les variables sont **sensibles à la casse** : `a` et `A` sont deux variables différentes

### Affectation
- L'affectation se fait avec le signe `=`
- Python permet l'**affectation multiple** : `a, b, c = 1, 2, 3`

> ⚠️ Les **constantes** n'existent pas nativement en Python. Par convention, on utilise des noms en MAJUSCULES (ex : `PI = 3.14`).

## Déclaration simple

Les variables peuvent contenir des valeurs de différents types. Pour déclarer une variable, on utilise le nom suivi du signe `=` et de la valeur :

```python
ma_variable = 1
ma_variable_2 = "a"
ma_variable_3 = True
```

## Affectation multiple

On peut affecter plusieurs valeurs à plusieurs variables en une seule ligne :

```python
a, b, c = 1, 2, 3
j,k,l = "a", "b", "c"
```

## Les types de base

Python est un langage à **typage dynamique** : le type d'une variable est déterminé automatiquement par la valeur qu'elle contient.

```python
# Entier (int)
value_int = 1

# Flottant (float) — le point est le séparateur décimal (norme internationale)
value_float = 1.0

# Chaîne de caractères (str)
value_string = "a"

# Valeur Unicode — permet des caractères spéciaux (accents, emojis, idéogrammes...)
value_unicode = "\U0001F44B"

# Booléen (bool) — True ou False
value_bool = True

# Chaîne multiligne — avec triple guillemets
value_multiline = """Ligne 1
Ligne 2
Ligne 3
"""

# Chaîne formatée (f-string) — insère des variables dans le texte
message = "chimichanga"
value_fstring = f"Hello {message}"

# Chaîne brute (raw string) — n'interprète pas les caractères spéciaux (\n, \t, etc.)
adresse = r"C:\Users\user\Documents\file.txt"

# None — représente l'absence de valeur
value_none = None

# type() — renvoie le type d'une variable
value_inconnu = 5
print(type(value_inconnu))  # Affiche <class 'int'>
```

## Vérifier le type d'une variable

La fonction `type()` permet de connaître le type d'une variable. Le résultat est affiché sous la forme `<class 'type'>` — les chevrons `<>` indiquent que c'est une référence à une classe.

```python
value_test = 5
print(type(value_test))

# affiche <class 'int'>, 
# ce qui signifie que value_test est de type entier
#les chevrons <> indiquent que c'est une classe
#Le format <...> permet de distinguer rapidement l'information d’un simple texte.
```

## Interaction avec la console

### `print()` — Afficher une valeur
Affiche du texte ou la valeur d'une variable dans la console.

### `input()` — Récupérer une saisie utilisateur
Demande à l'utilisateur de saisir une valeur (toujours de type `str`).

```python
# afficher la valeur d'une variable dans la console
value_to_display = "Hello world"
print(value_to_display)


#recuperer une valeur dans la console
input_value = input("Entrez une valeur : ") 


# Demande à l'utilisateur de saisir une valeur
text = input("Entrez une valeur : ")

text = text + " hello "
print(text)
```

### Arguments de `print()`

La fonction `print()` accepte des arguments supplémentaires :
- **`sep`** : définit le séparateur entre les valeurs (par défaut `" "`)
- **`end`** : définit le caractère de fin de ligne (par défaut `"\n"`)

```python
nb = 5
print("le nombre est", nb, sep=" - ", end="!")
```

## Conversion de types (casting)

Tout ce que `input()` renvoie est de type `str`. Pour effectuer des calculs, il faut souvent **convertir** la valeur vers un autre type.

Les fonctions de conversion sont :
- **`int()`** : convertit en entier
- **`float()`** : convertit en flottant
- **`str()`** : convertit en chaîne de caractères
- **`bool()`** : convertit en booléen

> On utilise le nom du type cible comme fonction de conversion.

```python
# int() : convertit en entier
value_str = input("Entrez un entier : ")
value_int = int(value_str)

# float() : convertit en flottant
value_str_2 = input("Entrez un décimal : ")
value_float = float(value_str_2)

# str() : convertit en chaîne de caractères
value_int = int(input("Entrez un nombre : "))
value_str = str(value_int)

# bool() : convertit en booléen
# Attention : bool("") → False, bool("n'importe quoi") → True
value_str_3 = input("Entrez une valeur : ")
value_bool = bool(value_str_3)

# Note : unicode() n'existe plus depuis Python 3
# Toutes les chaînes sont déjà en Unicode par défaut
```
