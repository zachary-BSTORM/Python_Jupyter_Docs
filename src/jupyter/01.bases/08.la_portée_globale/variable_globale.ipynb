# Les variables globales

Une variable **globale** est définie en dehors de toute fonction. Elle est accessible **partout** dans le fichier.

### Le mot-clé `global`

Pour **modifier** une variable globale depuis l'intérieur d'une fonction, il faut utiliser le mot-clé `global`. Sans cela, Python crée une nouvelle variable locale du même nom.

> ⚠️ L'utilisation de `global` est **déconseillée** en pratique. Préférez passer les valeurs en **paramètre** et les récupérer via `return`.

```python
compteur = 0

def incrementer():
    global compteur  # Sans ce mot-clé, Python créerait une variable locale
    compteur += 1

incrementer()
incrementer()
print(compteur)  # Affiche 2
```

### Meilleure pratique : passer en paramètre

Plutôt que d'utiliser `global`, il est préférable de passer la valeur en paramètre et de la retourner :

```python
def incrementer(valeur):
    return valeur + 1

compteur = 0
compteur = incrementer(compteur)
compteur = incrementer(compteur)
print(compteur)  # Affiche 2
```
