# Les variables locales

Une variable **locale** est définie à l'intérieur d'une fonction. Elle n'existe **que** pendant l'exécution de cette fonction et n'est **pas accessible** en dehors.

### Règles de portée

- Une variable créée dans une fonction est **locale** par défaut
- Elle est **détruite** à la fin de l'exécution de la fonction
- Deux fonctions peuvent utiliser le **même nom** de variable sans conflit

```python
def ma_fonction():
    message = "Je suis locale"  # Variable locale
    print(message)

ma_fonction()
# print(message)  # NameError : message n'existe pas en dehors de la fonction
```

### Priorité locale sur globale

Si une variable locale a le **même nom** qu'une variable globale, la version **locale** est utilisée à l'intérieur de la fonction :

```python
nom = "Global"

def saluer():
    nom = "Local"  # Variable locale, ne modifie pas la globale
    print(f"Dans la fonction : {nom}")

saluer()              # Affiche "Dans la fonction : Local"
print(f"Hors fonction : {nom}")  # Affiche "Hors fonction : Global"
```
