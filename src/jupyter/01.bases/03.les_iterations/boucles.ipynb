# Les itérations (boucles)

Les boucles permettent de **répéter** des instructions un certain nombre de fois ou tant qu'une condition est remplie.

La condition doit toujours renvoyer un **booléen** (`True` ou `False`).

Python propose deux types de boucles :
- **`while`** : répète tant qu'une condition est vraie
- **`for`** : parcourt une séquence d'éléments

## La boucle `while`

La boucle `while` exécute un bloc de code **tant que** la condition est vraie.

> ⚠️ Attention à toujours prévoir une **condition d'arrêt**, sinon la boucle tourne à l'infini !

```python
i = 0
while i < 10:
    print(i)
    i += 1
```

## La boucle `for`

La boucle `for` parcourt un **itérable** (liste, tuple, dictionnaire, chaîne de caractères, etc.).

### 1. Parcourir une collection directement

```python
collection = [1, 2, 3, 4, 5]
for value in collection:
    print(value)
```

### 2. Utiliser `range()` pour générer une séquence d'entiers

```python
for i in range(10):
    print(i)
```

### Paramètres de `range()`

| Paramètre | Description | Valeur par défaut |
|---|---|---|
| `start` | Premier entier de la séquence | `0` |
| `stop` | Dernier entier (**non inclus**) | — |
| `step` | Pas entre chaque entier | `1` |

```python
for i in range(1, 10, 2):
    print(i)
```

### 3. Utiliser `enumerate()` pour obtenir l'index et la valeur

`enumerate()` associe un **index** à chaque élément lors du parcours :

```python
for i, value in enumerate(collection):
    print(i, value)
```

## Contrôle de boucle : `break` et `continue`

| Mot-clé | Description |
|---|---|
| `break` | **Interrompt** la boucle immédiatement |
| `continue` | **Saute** l'itération en cours et passe à la suivante |

```python
# break : arrête la boucle quand on trouve 3
for i in range(10):
    if i == 3:
        break
    print(i)  # Affiche 0, 1, 2

print("---")

# continue : saute les nombres pairs
for i in range(6):
    if i % 2 == 0:
        continue
    print(i)  # Affiche 1, 3, 5
```
