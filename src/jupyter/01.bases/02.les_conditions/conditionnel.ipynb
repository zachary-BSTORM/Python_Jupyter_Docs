# Les structures conditionnelles

En Python, on n'utilise **pas d'accolades** pour délimiter les blocs de code, mais l'**indentation** (généralement 4 espaces). Un bloc commence après les **deux points** `:`.

Les structures conditionnelles permettent d'exécuter du code **en fonction d'une condition**.

### Les mots-clés

| Mot-clé | Description |
|---|---|
| `if` | Si la condition est **vraie**, le bloc est exécuté |
| `elif` | Si la condition précédente est fausse, on teste cette nouvelle condition |
| `else` | Si **aucune** des conditions précédentes n'est vraie |
| `match` | Évalue la valeur d'une variable (disponible depuis **Python 3.10**, équivalent du `switch` d'autres langages) |

> Les parenthèses autour de la condition sont **optionnelles** mais peuvent améliorer la lisibilité.

```python
nb = 10

# --- if seul ---
if nb > 0:
    print("Le nombre est positif")

# --- if / elif --- autant de elif que nécéssaire
if nb > 0:
    print("Le nombre est positif")
elif nb < 0:
    print("Le nombre est négatif")

# --- if / elif / else --- un seul else
if nb > 0:
    print("Le nombre est positif")
elif nb < 0:
    print("Le nombre est négatif")
else:
    print("Le nombre est zéro")
```

## Le `match / case` (Python 3.10+)

Équivalent du `switch` dans d'autres langages, le `match` permet de comparer une valeur à plusieurs cas possibles :

```python
commande = "quitter"

match commande:
    case "démarrer":
        print("Démarrage du programme")
    case "pause":
        print("Mise en pause")
    case "quitter":
        print("Fermeture du programme")
    case _:
        print("Commande inconnue")  # _ est le cas par défaut
```
