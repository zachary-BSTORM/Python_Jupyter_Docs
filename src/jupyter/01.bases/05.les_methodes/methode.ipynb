# Les fonctions et procédures

Une **fonction** permet d'encapsuler un bloc de code réutilisable.  
Au lieu de réécrire la même logique 10 fois, on l'écrit une fois dans une fonction et on l'appelle quand on en a besoin.

On déclare une fonction avec le mot-clé **`def`**.

| Type | Description | Mot-clé |
|---|---|---|
| **Fonction** | Renvoie une valeur | `return` |
| **Procédure** | Effectue une action, ne renvoie rien | Pas de `return` |

> En Python, toute fonction sans `return` retourne automatiquement `None` (= rien, vide).

````python
# Procédure : effectue une action, ne retourne rien
def afficher_bonjour():
    print("Bonjour !")


# Fonction : calcule et retourne une valeur
def additionner(a, b):
    return a + b


# Appel des fonctions
afficher_bonjour()          # affiche "Bonjour !"
resultat = additionner(3, 4)  # resultat = 7
print(resultat)


# Une procédure retourne None implicitement
valeur = afficher_bonjour()  # affiche "Bonjour !"
print(valeur)                # None
````

## Paramètres et valeurs par défaut

Une fonction peut recevoir des **paramètres** (des valeurs qu'on lui passe au moment de l'appel).  
On peut leur donner une **valeur par défaut** : si l'appelant ne fournit pas ce paramètre, la valeur par défaut est utilisée.

### Règle importante
Les paramètres **avec valeur par défaut** doivent toujours être placés **après** les paramètres obligatoires.

````python
def saluer(nom, message="Bonjour"):
    print(f"{message}, {nom} !")

saluer("Alice")               # Bonjour, Alice !  (valeur par défaut utilisée)
saluer("Bob", "Bonsoir")      # Bonsoir, Bob !    (valeur par défaut ignorée)


# On peut aussi passer les arguments par nom (plus lisible)
saluer(nom="Charlie", message="Salut")
saluer(message="Hey", nom="Diana")   # l'ordre n'a plus d'importance si on nomme


# ⚠️ Piège classique avec les valeurs par défaut mutables
# (une liste, un dict... = objet qui peut changer)
# La valeur par défaut est créée UNE seule fois, au moment où Python lit
# la ligne `def` (c'est-à-dire à l'import du fichier) — PAS à chaque appel.
# → elle est donc partagée entre tous les appels !

def ajouter_mauvais(valeur, liste=[]):   # ← MAUVAISE PRATIQUE
    liste.append(valeur)
    return liste

print(ajouter_mauvais(1))   # [1]
print(ajouter_mauvais(2))   # [1, 2]  ← la liste n'a pas été réinitialisée !


# BONNE PRATIQUE : utiliser None comme valeur par défaut
def ajouter_correct(valeur, liste=None):
    if liste is None:
        liste = []           # nouvelle liste créée à chaque appel
    liste.append(valeur)
    return liste

print(ajouter_correct(1))   # [1]
print(ajouter_correct(2))   # [2]  ← correct
````

## `*args` et `**kwargs` — Nombre variable d'arguments

Parfois on veut qu'une fonction accepte **autant d'arguments qu'on veut**, sans savoir à l'avance combien il y en aura.

| Syntaxe | Ce que ça reçoit | Type reçu dans la fonction |
|---|---|---|
| `*args` | Plusieurs arguments positionnels (sans nom) | `tuple` |
| `**kwargs` | Plusieurs arguments nommés (`clé=valeur`) | `dict` |

> **`args`** et **`kwargs`** sont juste des conventions de nommage — on pourrait les appeler `*nombres` ou `**options`.  
> Mais tout le monde utilise `*args` / `**kwargs`, donc reste sur ces noms.

**L'ordre obligatoire dans une signature :**
```
def f(obligatoires, *args, keyword_only, **kwargs)
```

````python
# *args — reçoit un tuple de tous les arguments positionnels supplémentaires
def somme(*args):
    print(args)           # (1, 2, 3) — c'est un tuple
    return sum(args)

print(somme(1, 2, 3))         # 6
print(somme(1, 2, 3, 4, 5))   # 15
print(somme())                # 0


# **kwargs — reçoit un dict de tous les arguments nommés supplémentaires
def afficher_infos(**kwargs):
    print(kwargs)         # {'nom': 'Alice', 'age': 30} — c'est un dict
    for cle, valeur in kwargs.items():
        print(f"  {cle} = {valeur}")

afficher_infos(nom="Alice", age=30, ville="Paris")

# Combinaison des deux
def fonction_complete(param1, *args, **kwargs):
    print(f"param1 = {param1}")
    print(f"args   = {args}")    # tout ce qui est positionnel après param1
    print(f"kwargs = {kwargs}")  # tout ce qui est nommé

fonction_complete(1, 2, 3, 4, x=10, y=20)
# param1 = 1
# args   = (2, 3, 4)
# kwargs = {'x': 10, 'y': 20}
````

````python
# Déballage (unpacking) — l'inverse : passer une liste/dict à une fonction
# On utilise * pour déballer une liste, ** pour déballer un dict

def additionner(a, b, c):
    return a + b + c

valeurs = [1, 2, 3]
print(additionner(*valeurs))    # équivalent à additionner(1, 2, 3)

params = {"a": 10, "b": 20, "c": 30}
print(additionner(**params))    # équivalent à additionner(a=10, b=20, c=30)
````

## Le `return`

- `return` **sort immédiatement** de la fonction et renvoie une valeur
- Tout ce qui est après un `return` dans la même branche n'est **jamais exécuté**
- On peut retourner **plusieurs valeurs** d'un coup (Python les regroupe dans un tuple automatiquement)
- Un `return` **ne sort que de la fonction** — pas de la boucle qui l'a appelée

````python
# return simple
def valeur_absolue(x):
    if x < 0:
        return -x    # sort ici si x est négatif
    return x         # sinon, sort ici


# Retourner plusieurs valeurs
def min_max(liste):
    return min(liste), max(liste)   # retourne un tuple (minimum, maximum)

petit, grand = min_max([3, 1, 4, 1, 5, 9])
print(petit, grand)    # 1 9


# return ne sort PAS de la boucle qui appelle la fonction
def verifier(x):
    if x == 0:
        return "zéro"    # sort de verifier() seulement
    return "autre"

for i in range(3):
    print(verifier(i))  # la boucle continue, elle n'est pas affectée par le return
# autre
# zéro
# autre
````

## Portée des variables (Scope)

Une variable n'est pas forcément accessible partout dans le code.  
Sa **portée** définit depuis où elle est visible.

| Portée | Où est définie la variable | Accessible depuis |
|---|---|---|
| **Locale** | Dans une fonction | Cette fonction uniquement |
| **Globale** | En dehors de toute fonction | Partout dans le fichier |

- `global` : permet de **modifier** une variable globale depuis une fonction
- `nonlocal` : permet de modifier une variable d'une fonction **englobante** (cas des fonctions imbriquées)

> **Bonne pratique** : évite d'utiliser `global`. Passe plutôt la variable en paramètre et retourne le résultat — c'est plus propre et plus prévisible.

````python
ma_valeur = "globale"

def ma_fonction():
    ma_valeur = "locale"     # crée une variable locale — ne modifie PAS la globale
    print(ma_valeur)         # "locale"

ma_fonction()
print(ma_valeur)             # "globale" — inchangée


# global — modifier une variable globale (à éviter)
compteur = 0

def incrementer():
    global compteur          # sans ça, Python essaierait de créer une variable locale
    compteur += 1

incrementer()
print(compteur)  # 1


# nonlocal — modifier une variable d'une fonction englobante
def creer_compteur():
    n = 0
    def incrementer():
        nonlocal n           # accède au n de creer_compteur
        n += 1
        return n
    return incrementer

compte = creer_compteur()
print(compte())  # 1
print(compte())  # 2
print(compte())  # 3
````

## Fonctions lambda

Une **lambda** est une fonction anonyme (sans nom) écrite en une seule ligne.  
Elle est utile pour des petites opérations simples qu'on passe en argument à une autre fonction.

```python
lambda paramètres : expression
```

> Une lambda ne peut contenir qu'**une seule expression** — pas de boucle, pas de bloc `if/else` **multiligne**.  
> En revanche, le `if/else` **en une ligne** (appelé **expression ternaire** : une valeur *si* condition *sinon* une autre valeur) est permis, car c'est une expression :
> ```python
> parite = lambda x: "pair" if x % 2 == 0 else "impair"
> ```
> Si la logique est complexe, écris une vraie fonction avec `def`.

````python
# Ces deux écritures sont équivalentes
def carre(x):
    return x ** 2

list_values = [{name:"bob",age:30},{name:"boba",age:31},{name:"bobc",age:32}]

carre_lambda = lambda x: x ** 2

print(carre(5))         # 25
print(carre_lambda(5))  # 25


# Plusieurs paramètres
additionner = lambda a, b: a + b
print(additionner(3, 4))   # 7


# Usage le plus courant : comme argument de sorted(), sort(), map(), filter()
personnes = [("Alice", 30), ("Bob", 25), ("Charlie", 35)]


# Trier par âge (le 2ème élément de chaque tuple)
personnes.sort(key=lambda p: p[1])
print(personnes)   # [('Bob', 25), ('Alice', 30), ('Charlie', 35)]

# Trier des mots sans tenir compte de la casse
mots = ["Python", "java", "C", "rust"]
tries = sorted(mots, key=lambda m: m.lower())
print(tries)       # ['C', 'java', 'Python', 'rust']

# test 
result = True if ("Alice",30) in personnes else False
print(result)
````

## `map()` et `filter()` — transformer et filtrer

Ces deux fonctions built-in (= intégrées à Python, pas besoin d'import) permettent de travailler sur des listes de façon concise.

| Fonction | Ce qu'elle fait |
|---|---|
| `map(fonction, liste)` | Applique `fonction` à **chaque élément** et retourne les résultats |
| `filter(fonction, liste)` | Garde uniquement les éléments pour lesquels `fonction` retourne `True` |

> `map()` et `filter()` retournent un **itérateur** (objet paresseux qui ne calcule que quand on en a besoin).  
> Pour voir les valeurs, convertir avec `list()`.  
> **Itérateur** : objet qu'on ne peut parcourir qu'une fois, élément par élément. Économise la mémoire.

````python
nombres = [1, 2, 3, 4, 5]

# map — appliquer une transformation à chaque élément
carres = list(map(lambda x: x**2, nombres))
print(carres)   # [1, 4, 9, 16, 25]

# Équivalent avec une list comprehension (souvent préféré car plus lisible)
carres2 = [x**2 for x in nombres]


# filter — garder seulement certains éléments
pairs = list(filter(lambda x: x % 2 == 0, nombres))
print(pairs)    # [2, 4]

# Équivalent avec une list comprehension
pairs2 = [x for x in nombres if x % 2 == 0]


# filter(None, liste) → garde les éléments "vrais" (élimine 0, None, "", [], False)
mixed = [0, 1, "", "ok", None, 42, False, []]
vrais = list(filter(None, mixed))
print(vrais)    # [1, 'ok', 42]
````

## Annotations de type (type hints)

Les annotations permettent d'indiquer dans le code quels **types** sont attendus pour les paramètres et la valeur de retour.  
Python **ne les vérifie pas** à l'exécution — elles servent uniquement de documentation et d'aide pour l'IDE (qui peut alors signaler des erreurs).

```python
def nom(param: type) -> type_retour:
```

| Annotation | Signification |
|---|---|
| `x: int` | x est un entier |
| `x: str` | x est une chaîne |
| `x: list` | x est une liste |
| `x: int \| None` | x est un int **ou** None (Python 3.10+) |
| `-> str` | la fonction retourne une chaîne |
| `-> None` | la fonction ne retourne rien |

````python
# Annotations simples
def saluer(nom: str, fois: int = 1) -> str:
    return ("Bonjour " + nom + " ") * fois

print(saluer("Alice", 2))   # Bonjour Alice  Bonjour Alice


# int | None → le paramètre peut être un int ou None
# (Python 3.10+, avant il fallait écrire Optional[int] depuis le module typing)
def afficher(valeur: int | None = None) -> None:
    if valeur is None:
        print("Aucune valeur")
    else:
        print(valeur)


# Avec *args et **kwargs
def journaliser(*messages: str, niveau: str = "INFO") -> None:
    for msg in messages:
        print(f"[{niveau}] {msg}")

journaliser("Démarrage", "Connexion", niveau="DEBUG")
````

## Imports utiles pour les fonctions — performances

Python possède plusieurs modules de la bibliothèque standard (= livrés avec Python, pas besoin d'installer quoi que ce soit) qui améliorent le travail avec les fonctions.

### Le module `functools`

**`functools`** contient des outils pour travailler avec des fonctions :

| Outil | Utilité |
|---|---|
| `@cache` | Mémoïze les résultats (évite de recalculer la même chose) |
| `@lru_cache(maxsize=N)` | Comme `@cache` mais avec une limite de taille |
| `@wraps` | Préserve le nom et la doc d'une fonction décorée |
| `reduce()` | Réduit une liste à une seule valeur |
| `partial()` | Fixe certains arguments d'une fonction |

> **Mémoïzation** : technique qui consiste à **sauvegarder le résultat** d'un calcul pour ne pas le refaire si on rappelle la fonction avec les mêmes arguments. C'est un compromis : on utilise plus de mémoire pour aller plus vite.

````python
from functools import cache, lru_cache

# Sans cache — recalcule à chaque appel
def fibonacci_lent(n):
    if n <= 1:
        return n
    return fibonacci_lent(n - 1) + fibonacci_lent(n - 2)

# fibonacci_lent(35) → très lent (des millions d'appels récursifs)


# Avec @cache — retient les résultats déjà calculés
@cache
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(35))    # quasi instantané
print(fibonacci(100))   # toujours rapide


# @lru_cache(maxsize=N) — comme @cache mais limite le nombre de résultats gardés en mémoire
# LRU = Least Recently Used : quand la cache est pleine, on supprime le résultat le moins récemment utilisé
@lru_cache(maxsize=128)  # garde au maximum 128 résultats
def calcul_couteux(n):
    return sum(range(n))

print(calcul_couteux(1000))
print(calcul_couteux(1000))  # retourné instantanément depuis la cache

# Voir les stats de la cache
print(calcul_couteux.cache_info())
# CacheInfo(hits=1, misses=1, maxsize=128, currsize=1)
# hits   = nombre de fois où le résultat était déjà en cache
# misses = nombre de fois où le calcul a vraiment été fait

# Vider la cache si besoin
calcul_couteux.cache_clear()
````

### `reduce()` — réduire une liste à une valeur

`reduce(fonction, liste)` applique une fonction **cumulativement** sur les éléments d'une liste pour aboutir à une seule valeur.  

Par exemple `reduce(lambda acc, x: acc + x, [1,2,3,4])` fait : `((1+2)+3)+4` = `10`

> **acc** = accumulateur : la valeur cumulée jusqu'ici. **x** = l'élément courant.

````python
from functools import reduce

nombres = [1, 2, 3, 4, 5]

# Somme
total = reduce(lambda acc, x: acc + x, nombres)   # 15
print(total)

# Produit
produit = reduce(lambda acc, x: acc * x, nombres)  # 1*2*3*4*5 = 120
print(produit)

# Maximum sans utiliser max()
maximum = reduce(lambda acc, x: acc if acc > x else x, nombres)  # 5
print(maximum)

# Valeur initiale (3ème argument optionnel)
total2 = reduce(lambda acc, x: acc + x, nombres, 100)  # 100 + 1+2+3+4+5 = 115
print(total2)
````

### `partial()` — fixer des arguments à l'avance

`partial(fonction, arg1, arg2, ...)` crée une **nouvelle fonction** basée sur une existante, avec certains arguments déjà fixés.  
Utile quand tu dois appeler une fonction toujours avec les mêmes arguments répétitifs.

````python
from functools import partial

def puissance(base, exposant):
    return base ** exposant

# Créer une version spécialisée avec exposant déjà fixé à 2
carre = partial(puissance, exposant=2)
cube  = partial(puissance, exposant=3)

print(carre(5))   # 25
print(cube(3))    # 27


# Autre exemple : print avec séparateur fixé
print_virgule = partial(print, sep=", ")
print_virgule("Alice", "Bob", "Charlie")   # Alice, Bob, Charlie
````

## Autres modules utiles avec les fonctions

### `time` — mesurer les performances

Utile pour savoir combien de temps prend une fonction à s'exécuter.

````python
import time

def ma_fonction_lente():
    time.sleep(0.5)   # simule un calcul qui prend 0.5 seconde
    return 42

# Méthode simple : chronomètre manuel
debut = time.perf_counter()       # perf_counter = horloge haute précision
resultat = ma_fonction_lente()
fin = time.perf_counter()

print(f"Résultat : {resultat}")
print(f"Durée : {fin - debut:.4f} secondes")
````

### `timeit` — mesurer précisément

`timeit` est plus précis que `time` pour comparer des petites opérations : il exécute le code **plusieurs milliers de fois** et fait la moyenne, pour éliminer les variations aléatoires du système.

````python
import timeit

# Comparer deux façons de créer une liste

# Méthode 1 : list comprehension
temps1 = timeit.timeit(
    stmt="[x**2 for x in range(1000)]",  # le code à mesurer
    number=1000                           # nombre de répétitions
)

# Méthode 2 : map
temps2 = timeit.timeit(
    stmt="list(map(lambda x: x**2, range(1000)))",
    number=1000
)

print(f"List comprehension : {temps1:.4f}s")
print(f"map()              : {temps2:.4f}s")
print(f"→ La plus rapide est {'list comprehension' if temps1 < temps2 else 'map()'}")
````

### `itertools` — outils pour les itérations

`itertools` contient des fonctions pour créer et combiner des itérateurs efficacement.  
Tout est calculé à la demande (paresseux = économe en mémoire).

| Fonction | Ce qu'elle fait |
|---|---|
| `chain(*iterables)` | Enchaîne plusieurs itérables comme s'ils n'en formaient qu'un |
| `islice(iterable, stop)` | Prend les N premiers éléments d'un itérable |
| `enumerate` | Déjà disponible sans import |
| `product(*iterables)` | Produit cartésien (toutes les combinaisons) |
| `combinations(iterable, r)` | Toutes les combinaisons de r éléments (sans répétition) |
| `permutations(iterable, r)` | Toutes les permutations de r éléments |

````python
import itertools

# chain — enchaîner des listes
liste1 = [1, 2, 3]
liste2 = [4, 5, 6]
liste3 = [7, 8, 9]
tout = list(itertools.chain(liste1, liste2, liste3))
print(tout)   # [1, 2, 3, 4, 5, 6, 7, 8, 9]


# islice — prendre les N premiers éléments
premiers = list(itertools.islice(range(1000000), 5))
print(premiers)   # [0, 1, 2, 3, 4]  — sans charger le million en mémoire


# product — toutes les combinaisons possibles
couleurs = ["rouge", "bleu"]
tailles  = ["S", "M", "L"]
produits = list(itertools.product(couleurs, tailles))
print(produits)
# [('rouge', 'S'), ('rouge', 'M'), ('rouge', 'L'), ('bleu', 'S'), ...]


# combinations — choisir r éléments parmi n (l'ordre ne compte pas)
equipes = list(itertools.combinations(["Alice", "Bob", "Charlie", "Diana"], 2))
print(equipes)
# [('Alice', 'Bob'), ('Alice', 'Charlie'), ('Alice', 'Diana'), ...]


# permutations — toutes les façons d'ordonner r éléments
podium = list(itertools.permutations(["Alice", "Bob", "Charlie"], 3))
print(f"{len(podium)} podiums possibles")
````

# Demo quick sort

- division sur base de la dernière valeur 
- on récupère ce qui est plus grand et ce qui est plus petit et on recommence 

````python
import random

def quick_sort(liste: list) -> list:
    if len(liste) <= 1:
        return liste
    pivot = random.choice(liste)
    petit = [v for v in liste if v < pivot]
    egal  = [v for v in liste if v == pivot]
    grand = [v for v in liste if v > pivot]
    return quick_sort(petit) + egal + quick_sort(grand)


values = [5,99,2,3,8,46,5,4,6,8,9,55,68,897,420,654,547,232,125]

print(quick_sort(values))
````
