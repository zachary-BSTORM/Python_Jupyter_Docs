```python
import matplotlib.pyplot as plt
import numpy as np

import matplotlib
import matplotlib as mpl


vegetables = ["cucumber", "tomato", "lettuce", "asparagus",
              "potato", "wheat", "barley"]
farmers = ["Farmer Joe", "Upland Bros.", "Smith Gardening",
           "Agrifun", "Organiculture", "BioGoods Ltd.", "Cornylee Corp."]

harvest = np.array([[0.8, 2.4, 2.5, 3.9, 0.0, 4.0, 0.0],
                    [2.4, 0.0, 4.0, 1.0, 2.7, 0.0, 0.0],
                    [1.1, 2.4, 0.8, 4.3, 1.9, 4.4, 0.0],
                    [0.6, 0.0, 0.3, 0.0, 3.1, 0.0, 0.0],
                    [0.7, 1.7, 0.6, 2.6, 2.2, 6.2, 0.0],
                    [1.3, 1.2, 0.0, 0.0, 0.0, 3.2, 5.1],
                    [0.1, 2.0, 0.0, 1.4, 0.0, 1.9, 6.3]])


fig, ax = plt.subplots()
im = ax.imshow(harvest)

# Show all ticks and label them with the respective list entries
ax.set_xticks(range(len(farmers)), labels=farmers,
              rotation=45, ha="right", rotation_mode="anchor")
ax.set_yticks(range(len(vegetables)), labels=vegetables)

# Loop over data dimensions and create text annotations.
for i in range(len(vegetables)):
    for j in range(len(farmers)):
        text = ax.text(j, i, harvest[i, j],
                       ha="center", va="center", color="w")

ax.set_title("Harvest of local farmers (in tons/year)")
fig.tight_layout()
plt.show()
```

# Types de graphiques

| Type | Description | Méthode |
|---|---|---|
| Line plot | Courbe de données continues | `plt.plot()` |
| Bar chart | Comparaison de catégories | `plt.bar()` |
| Barh | Barres horizontales | `plt.barh()` |
| Histogram | Distribution de données | `plt.hist()` |
| Scatter plot | Relation entre deux variables | `plt.scatter()` |
| Pie chart | Proportions de catégories | `plt.pie()` |
| Box plot | Répartition statistique | `plt.boxplot()` |
| Area plot | Courbe avec remplissage | `plt.fill_between()` |
| Stacked bar | Barres empilées | `plt.bar(..., bottom=...)` |
| Heatmap | Représentation matricielle | `plt.imshow()` |
