# Tracer des courbes avec Matplotlib

**Matplotlib** est packagé avec Pandas et directement accessible.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('classic')
df = pd.read_csv("../../03.numpy_panda/pandas/data.csv")
````

````python
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x), '-')
plt.plot(x, np.cos(x), '--')
plt.show()

# np.sin(x) : calcule le sinus de l’angle x (en radians).

# np.cos(x) : calcule le cosinus de l’angle x (en radians).
````

## Personnaliser les couleurs

Chaque ligne tracée peut avoir sa propre couleur. Matplotlib accepte plusieurs formats de spécification :

| Format | Exemple |
|---|---|
| Nom | `color='blue'` |
| Code court | `color='g'` |
| Noir/blanc (0-1) | `color='0.75'` |
| Hex | `color='#FFDD44'` |
| RGB tuple | `color=(1.0, 0.2, 0.3)` |

````python
plt.plot(x, np.sin(x - 0), color='blue') # par nom
plt.plot(x, np.sin(x - 1), color='g') # par code
plt.plot(x, np.sin(x - 2), color='0.75') # noir et blanc (0-1)
plt.plot(x, np.sin(x - 3), color='#FFDD44') # Hex code (RRGGBB [00-FF])
plt.plot(x, np.sin(x - 4), color=(1.0,0.2,0.3)) # RVB tuple (0-1)
plt.plot(x, np.sin(x - 5), color='chartreuse');
````

````python
plt.plot(x, x + 4, linestyle='--', color='blue') 
````

## Contrôler les limites des axes

Il est possible de **contrôler la zone visible** du graphique avec les méthodes suivantes :

- **`plt.xlim(min, max)`** : définit les limites de l'axe X
- **`plt.ylim(min, max)`** : définit les limites de l'axe Y
- **`plt.axis([xmin, xmax, ymin, ymax])`** : définit les quatre limites en une seule commande

````python
plt.plot(x, np.sin(x))
plt.xlim(-1, 11)           # Limite de l’axe des x
plt.ylim(-1.5, 1.5)        # Limite de l’axe des y

#la même chose, en une seule ligne : 
#plt.axis([xmin, xmax, ymin, ymax])
````

## Tracer avec `plot()`

La méthode **`plot()`** est la fonction principale pour afficher un graphique. Avec **Pandas**, on peut directement appeler `.plot()` sur un **DataFrame** ou une **Series** pour tracer les données sans passer par `plt.plot()` manuellement.

````python
# conversion du type string en float
df["price_paid"] = df["price_paid"].str.replace("$", "", regex=False).astype(float)

df.groupby("date")["price_paid"].sum().plot()
````

````python
# on peut modifier la taille de schema

df.groupby("date")["price_paid"].sum().plot(figsize=(20, 10))
````
