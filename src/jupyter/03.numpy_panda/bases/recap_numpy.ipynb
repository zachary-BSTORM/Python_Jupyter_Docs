# Récap des règles avec NumPy

1. **Type homogène** : tous les éléments d'un `ndarray` ont le même type
2. **Vue vs copie** : le slicing crée une **vue** (view), pas une copie
3. **Compatibilité des shapes** : les formes doivent être compatibles pour les opérations
4. **Attributs essentiels** :

| Attribut | Description |
|----------|-------------|
| `shape`  | Dimensions du tableau (lignes, colonnes, …) |
| `ndim`   | Nombre de dimensions |
| `size`   | Nombre total d'éléments |
| `dtype`  | Type de données |

5. **Broadcasting** : NumPy étend automatiquement les dimensions pour permettre les opérations
6. **Mémoire contiguë et performance** : les opérations vectorisées sont plus rapides que des boucles
7. **Reshape** : les dimensions doivent être compatibles pour le `reshape`
8. **Indexation booléenne** : `print(a[a > 25])`
9. **Opérations élément par élément**
10. **Transposition ≠ reshape** : `.T` inverse lignes ↔ colonnes, mais ne change pas la structure mémoire

## np.array()

Créer un tableau NumPy à partir d'une liste Python.

````python
import numpy as np
a = np.array([1, 2, 3])
print(a)  # [1 2 3]
````

## np.arange(start, stop, step)

Créer un tableau d'entiers espacés régulièrement (comme `range`).

````python
a = np.arange(1, 10, 2)
print(a)  # [1 3 5 7 9]
````

## np.linspace(start, stop, num)

Créer un tableau de **num** valeurs espacées régulièrement entre **start** et **stop**.

````python
a = np.linspace(0, 1, 5)
print(a)  # [0.   0.25 0.5  0.75 1.  ]
````

## np.zeros(shape) / np.ones(shape)

Créer un tableau rempli de **0** ou de **1**.

````python
print(np.zeros((2, 3)))  # 2 lignes, 3 colonnes de zéros
print(np.ones((2, 2)))   # 2 lignes, 2 colonnes de uns
````

## np.eye(n)

Créer une **matrice identité** (diagonale à 1).

````python
print(np.eye(3))
# [[1. 0. 0.]
#  [0. 1. 0.]
#  [0. 0. 1.]]
````

## reshape(new_shape)

Changer la **forme** d'un tableau sans copier les données.

````python
a = np.arange(9)
b = a.reshape((3, 3))
print(b)
````

## Attributs : shape, size, ndim

Obtenir des informations sur un tableau :

- `.shape` → dimensions du tableau
- `.size` → nombre total d'éléments
- `.ndim` → nombre de dimensions

## np.random.randint(low, high, size)

Générer des **entiers aléatoires** entre `low` (inclus) et `high` (exclus).

````python
a = np.random.randint(0, 10, size=(2, 3))
print(a)  # Par exemple : [[2 6 1] [9 0 7]]
````

## ndarray.T — Transposer

**Transposer** un tableau : inverser lignes et colonnes.

````python
a = np.array([[1, 2], [3, 4]])
print(a.T)
# [[1 3]
#  [2 4]]
````

## np.concatenate()

Coller deux tableaux ensemble (en ligne avec `axis=0` ou en colonne avec `axis=1`).

````python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])
print(np.concatenate((a, b), axis=0))  # Ajout en ligne
````

## np.hstack() / np.vstack()

Empiler des tableaux **horizontalement** (`hstack`) ou **verticalement** (`vstack`).

````python
a = np.array([1, 2])
b = np.array([3, 4])
print(np.hstack((a, b)))  # [1 2 3 4]
print(np.vstack((a, b)))  # [[1 2]
                          #  [3 4]]
````

## Fonctions statistiques

`np.mean()`, `np.sum()`, `np.min()`, `np.max()`, `np.std()` — statistiques simples sur un tableau.

````python
a = np.array([1, 2, 3, 4])
print(np.mean(a))  # 2.5
print(np.sum(a))   # 10
print(np.std(a))   # 1.118...
````

## np.where(condition)

Obtenir les **indices** où une condition est vraie.

````python
a = np.array([1, 3, 5, 2, 4])
print(np.where(a > 3))  # (array([2, 4]),)
````

## np.unique()

Trouver les **valeurs uniques** d'un tableau.

````python
a = np.array([1, 2, 2, 3, 3, 3])
print(np.unique(a))  # [1 2 3]
````

## np.dot(a, b) — Produit matriciel

**Produit matriciel** entre deux tableaux. Syntaxe alternative : `a @ b`.

````python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(np.dot(a, b))  # ou a @ b
````

## Transposition en détail

`.T` retourne la **transposée** du tableau : il inverse les axes.

- Les **lignes** deviennent des **colonnes**
- Les **colonnes** deviennent des **lignes**
- Pour un tableau 2D de shape `(m, n)`, la transposition donne un tableau `(n, m)`

> **Attention** : `.T` ne s'applique pas sur les tableaux 1D (aucun effet). Sur les tableaux 3D, il inverse tous les axes.

````python
# x.T


x = np.array([[1, 2, 3],
              [4, 5, 6]])

print("Original :\n", x)
print("Transposé :\n", x.T)
````

## Transformer un vecteur 1D en matrice

**Objectif** : transformer un vecteur 1D en **matrice ligne** ou **matrice colonne**.

Un vecteur 1D est juste une ligne d'éléments à plat, avec un shape de type `(n,)`. Pour obtenir une matrice 2D, il faut préciser le nombre de lignes et de colonnes.

````python
x = np.array([1, 2, 3])
print(x.shape)  # (3,) → une seule ligne, pas de colonne

# Mais pour faire des matrices (tableaux 2D), il faut dire explicitement :

# Combien de lignes ?

# Combien de colonnes ?
````

### En matrice ligne

Transformer `x = [1, 2, 3]` en `[[1, 2, 3]]` → shape `(1, 3)`

````python
# Pour obtenir ceci : 
[[1, 2, 3]]   # 1 ligne, 3 colonnes
````

**Solution 1** : `reshape((1, 3))` — je veux 1 ligne, 3 colonnes

````python
x.reshape((1, 3))

# [[1 2 3]]
````

**Solution 2** : `x[np.newaxis, :]` — ajoute un nouvel axe en ligne

````python
x[np.newaxis, :]  # On garde toutes les colonnes (:) et on ajoute une ligne

# [[1 2 3]]
````

### En matrice colonne

Transformer `x = [1, 2, 3]` en `[[1], [2], [3]]` → shape `(3, 1)`

On veut obtenir un tableau 2D avec 3 lignes et 1 colonne :

```
[[1],
 [2],
 [3]]
```

**Solution 1** : `reshape((3, 1))` — je veux 3 lignes, 1 colonne

````python
x.reshape((3, 1))

# [[1]
#  [2]
#  [3]]
````

**Solution 2** : `x[:, np.newaxis]` — je garde toutes les lignes et j'ajoute un nouvel axe en colonne

````python
x[:, np.newaxis]

# [[1]
#  [2]
#  [3]]
````

### Résumé

| Opération | Syntaxe | Résultat | Shape |
|-----------|---------|----------|-------|
| Vecteur ligne | `x.reshape(1, 3)` | `[[1, 2, 3]]` | `(1, 3)` |
| Vecteur ligne | `x[np.newaxis, :]` | `[[1, 2, 3]]` | `(1, 3)` |
| Vecteur colonne | `x.reshape(3, 1)` | `[[1], [2], [3]]` | `(3, 1)` |
| Vecteur colonne | `x[:, np.newaxis]` | `[[1], [2], [3]]` | `(3, 1)` |

## np.concatenate — Détails

`np.concatenate` permet de **coller plusieurs tableaux** ensemble.

Le paramètre **axis** indique comment on les colle :

- `axis=0` → on colle **par en dessous** (ligne après ligne)
- `axis=1` → on colle **par le côté** (colonne après colonne)

> Les dimensions doivent être compatibles pour l'axe non utilisé.

````python
np.concatenate([x, x], axis=0)
````
