# NumPy - Introduction

**NumPy** (Numerical Python) est une librairie qui permet de charger, stocker et manipuler de manière efficace des données avec Python.

Les données peuvent prendre beaucoup de formes : documents texte, images, son, mesures topographiques, sismiques, etc.

Malgré cette hétérogénéité, nous pouvons souvent considérer les données comme des **tableaux de chiffres**. Par exemple, les images en noir et blanc sont des tableaux en deux dimensions avec des pixels et une valeur associée comprise entre 0 et 255.

Les tableaux (**array**) dans NumPy ressemblent aux listes de Python mais ils sont beaucoup plus optimisés.

> NumPy est une librairie fondamentale pour traiter des données en Python.

## Typage des données en Python

Comment NumPy améliore-t-il la gestion des données en Python ?

Python est un langage à **typage dynamique** (≠ typage statique comme C ou Java).

Ceci a pour conséquence que les variables en Python stockent plus que leur valeur.

En effet, en plus de la valeur associée à la variable, il faut stocker le **type** de chaque valeur.

````python
x = 5
x = "hello"
````

## Structure d'un integer en Python

L'implémentation de Python est écrite en C. Tous les objets Python ne sont rien d'autres que des **structures C** bien déguisées.

Par exemple, si on définit un `int` en Python comme suit :

```python
x = 10000
```

`x` n'est pas un simple integer. Un seul integer contient en réalité **quatre parties** :

- **`ob_refcnt`** : permet de gérer l'allocation de mémoire
- **`ob_type`** : encode le type de la variable
- **`ob_size`** : spécifie la taille des membres
- **`ob_digit`** : contient la valeur de la variable

> Contrairement au langage C, les integers en Python possèdent une en-tête qui alourdit chaque valeur.

## Qu'est-ce qu'un NumPy Array ?

Quand toutes les variables sont d'un **même type**, une partie de l'information est redondante.

La différence entre une liste Python à typage dynamique et un NumPy array réside dans le stockage : les NumPy arrays perdent en **flexibilité** mais gagnent en **taille** et donc en **efficacité**.

## Créer un NumPy Array

````python
import numpy as np
````

````python
np.array([1, 4, 2, 5, 3])
````

### Conversion de type automatique

Les NumPy arrays ne peuvent contenir qu'un **seul type**. Si ce n'est pas le cas, NumPy va tenter une conversion (ici, les integers sont convertis en float) :

````python
np.array([3.14, 4, 2, 3])
````

### Spécifier le type avec `dtype`

On peut spécifier le type de données lorsque l'on crée un array avec le paramètre **`dtype`** :

````python
np.array([1, 2, 3, 4], dtype='float32')
````

### Tableaux multidimensionnels

Contrairement aux listes Python, les NumPy arrays peuvent être explicitement **multidimensionnels** :

````python
# Des listes imbriquées donnent un numpy array en plusieurs dimensions
np.array([range(i, i + 3) for i in [2, 4, 6]])
````

### Créer des arrays depuis zéro

````python
# Permet de créer un array remplit de 10 integers avec la valeur 0
np.zeros(10, dtype=int)

# Crée una matrice 3x5 remplie de 1 en float
np.ones((3, 5), dtype=float)

# Crée une matrice 3x5 remplie de 3,14
np.full((3, 5), 3.14)

# Crée un numpy array avec une séquence linéaire
# Commence à 0 jusqu’à 20 par step de 2
# Similaire à la fonction range
np.arange(0, 20, 2)

# Crée une matrice avec 5 nombres équidistants entre 0 et 1 (gateau)
np.linspace(0, 1, 5)

# Crée une matrice 3x3 de nombres aléatoires
np.random.random((3, 3)) # Comment faire pour voir toutes les distributions proposées par NumPy?

# Crée une matrice 3x3 suivant une distribution normale suivant les paramètres:
# d’une moyenne de 0 et d’écart type de 1
np.random.normal(0, 1, (3, 3))
````

## Types de données NumPy

| Type | Description |
|------|-------------|
| `bool_` | Booléen (`True` ou `False`) — stocké sur 1 byte |
| `int_` | Entier par défaut dépendant de l'architecture (souvent équivalent à `int64`) |
| `intc` | Entier identique au `int` de C |
| `intp` | Entier utilisé pour l'indexation |
| `int8` | Entier signé 8 bits : de -128 à 127 |
| `int16` | Entier signé 16 bits : de -32 768 à 32 767 |
| `int32` | Entier signé 32 bits : de -2³¹ à 2³¹-1 |
| `int64` | Entier signé 64 bits : très grands entiers |
| `uint8` | Entier non signé 8 bits : de 0 à 255 |
| `uint16` | Entier non signé 16 bits : de 0 à 65 535 |
| `uint32` | Entier non signé 32 bits : de 0 à 4 294 967 295 |
| `uint64` | Entier non signé 64 bits : très grands entiers positifs |
| `float16` | Nombre à virgule flottante 16 bits (précision réduite) |
| `float32` | Nombre flottant 32 bits (simple précision) |
| `float64` | Nombre flottant 64 bits (double précision — par défaut) |
| `complex64` | Nombre complexe (2 × 32 bits = réel + imaginaire) |
| `complex128` | Nombre complexe (2 × 64 bits = réel + imaginaire) |

> `complex` est une abréviation de `complex128`.

## Attributs des NumPy Array

Chaque NumPy array possède des attributs utiles :

| Attribut | Description |
|----------|-------------|
| `ndim` | Nombre de dimensions |
| `shape` | Taille de chaque dimension |
| `size` | Nombre total d'éléments |
| `dtype` | Type des données |
| `itemsize` | Taille (en bytes) de chaque élément |
| `nbytes` | Taille totale (en bytes) du tableau |

````python
import numpy as np

np.random.seed(0) # seed pour la reproductibilité

x1 = np.random.randint(10, size=6) # array 1D

x2 = np.random.randint(10, size=(3, 4)) # array 2D

x3 = np.random.randint(10, size=(3, 4, 5)) # array 3D


print(x1)
print("x1 ndim: ", x1.ndim)
print()
print(x2)
print("x2 ndim: ", x2.ndim)
print()
print(x3)
print("x3 ndim: ", x3.ndim)
print("x3 shape:", x3.shape)
print("x3 size: ", x3.size)
print("dtype:", x3.dtype)
print("itemsize:", x3.itemsize, "bytes")
print("nbytes:", x3.nbytes, "bytes")
````

## Indexation des arrays

Accéder à un seul élément d'un NumPy array fonctionne de la même manière qu'avec les listes Python :

````python
# Positive indexing:
array_name = [1, 2, 3, 4, 5]

array_name[1] = 10  # Modifie le deuxième élément (index 1) de l'array 
print(array_name[1])  # Affiche 10
# 0 <= 2 < array_name.shape[1]
````

````python
# Negative indexing:

array_name[i]  

-array_name.shape[0] <= i <= -1
````

````python
# Array en 2D:

x2[0, 0]
````
