# Ajouter des colonnes

Il est possible d'**ajouter des colonnes** à un DataFrame, par exemple pour stocker le résultat d'un **calcul** basé sur des colonnes existantes.

Avant de créer une colonne calculée, il est souvent nécessaire de **nettoyer les données** (valeurs manquantes, types incorrects).

```python
import pandas as pd

# Chargement du fichier CSV
df = pd.read_csv("../data.csv")

# Remplissage des valeurs manquantes dans la colonne "tax" par 0
df["tax"] = df["tax"].fillna(0)  # Important d’assigner le résultat à la colonne

# Suppression du symbole "$" dans la colonne "price_paid" puis conversion en float
df["price_paid"] = df["price_paid"].apply(lambda x: x.replace("$", ""))
df["price_paid"] = df["price_paid"].astype(float)

# Vérification des types des colonnes
df.dtypes

```

## Syntaxe de base

Pour **ajouter** ou **supprimer** une colonne :

| Action | Syntaxe |
|---|---|
| Ajouter une colonne | `df["nouvelle_colonne"] = valeur` |
| Supprimer une colonne | `del df["colonne"]` |

## Créer une colonne calculée

Nous allons créer une colonne **`total_price`** en combinant le prix et la taxe.

```python
# Remplir les valeurs manquantes de la colonne "tax" avec 0
df["tax"] = df["tax"].fillna(0) # Important d’assigner le résultat à la colonne

# Calcul du prix total après application de la taxe
df["total_price"] = df["price_paid"] * (1 - df["tax"] / 100)


df
```

## Mapping avec `map()`

La méthode **`map()`** permet de **transformer** les valeurs d'une colonne à l'aide d'un **dictionnaire** de correspondance.

### Étapes :
1. Créer un dictionnaire `{"ancienne_valeur": "nouvelle_valeur"}`
2. Appliquer avec `df["col"].map(dictionnaire)`

> Les valeurs absentes du dictionnaire seront remplacées par `NaN`.

```python
countries = {"United States": "USA", "United Kingdom": "UK", "France": "FR", "Germany": "DE" , "Morocco": "MA" , "Canada" : "CA"}
# Remplacement des noms de pays par leurs abréviations

df["country_code"] = df["country"].map(countries)

df.head()
```
