# Analyser un DataFrame

````python
import pandas as pd

df = pd.read_csv('../data.csv')

df
````

## head()

Par défaut, affiche les **5 premières lignes**. On peut passer un nombre en paramètre : `df.head(10)`

````python
df.head()

#df.head(10)

#df.head(n = 10)

````

## tail()

Affiche les **dernières lignes** du DataFrame.

````python
df.tail()

# df.tail(3)

# df.tail(n = 10)
````

## shape

Attribut (sans parenthèses) qui retourne un **tuple** `(nombre de lignes, nombre de colonnes)`.

````python
df.shape

#df
````

## columns

Attribut qui retourne l'**index des colonnes** du DataFrame (objet Pandas).

On peut utiliser `tolist()` pour obtenir une **liste Python** classique :

```python
df.columns.tolist()
```

````python
df.columns
````

## index

Retourne les **index** (étiquettes des lignes) du DataFrame.

````python
df.index
````

## set_index()

Permet de **définir une colonne comme index** du DataFrame.

```python
df.set_index("nom_colonne")
```

| Paramètre | Type | Description |
|-----------|------|-------------|
| `inplace` | `bool` | Si `True`, modifie le DataFrame **en place**. Par défaut `False` (retourne une copie). |

````python
df_copie = df.set_index("email")

df_copie = df_copie.index

list(df_copie)
````
