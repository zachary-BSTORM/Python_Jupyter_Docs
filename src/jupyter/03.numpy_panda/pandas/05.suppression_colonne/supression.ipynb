# Supprimer des colonnes

La méthode **`drop()`** permet de supprimer des **lignes** ou des **colonnes** d'un DataFrame.

| Paramètre | Description |
|---|---|
| `axis=0` | Supprime des lignes |
| `axis=1` | Supprime des colonnes |
| `inplace=True` | Modifie le DataFrame en place |

````python
import pandas as pd

df = pd.read_csv("../data.csv")

df_test = df.drop("ip_address", axis=1)

# df.drop("ip_address", axis=1 , inplace=True) # permet de travailler sur la source  



df_test
````

## Supprimer plusieurs colonnes

**`drop()`** accepte une **liste** de noms de colonnes pour en supprimer plusieurs à la fois.

````python
df_copy = df.drop(["first_name","last_name","email"],axis=1)

df_copy
````

## Alternative : `del`

Le mot-clé **`del`** permet de supprimer une colonne directement sur le DataFrame original :

```python
del df["nom_colonne"]
```

> ⚠️ Contrairement à **`drop()`**, **`del`** modifie le DataFrame **en place** et ne retourne rien.

````python
del df["ip_address"]
````
