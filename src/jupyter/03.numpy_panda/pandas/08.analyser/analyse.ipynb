```python
import pandas as pd

df = pd.read_csv("../data.csv")
```

# Analyser les données

Maintenant nous allons pouvoir faire parler les données. On utilise la méthode **`describe()`** qui affiche un **tableau récapitulatif** avec des statistiques descriptives (moyenne, min, max, écart-type, etc.).

```python
df.describe()
```

```python
df["price_paid"].describe()
```

## Moyenne avec `mean()`

Nous pouvons récupérer la **moyenne** d'une colonne grâce à **`mean()`**.

> Ne pas oublier de faire une **conversion** si nécessaire (ex. : `str` → `float`).

```python

# mettre les valeurs manquantes à 0
df["tax"] = df["tax"].fillna(0)

# convertir les valeurs de la colonne "price_paid" en float
df["price_paid"] = df["price_paid"].apply(lambda x: x.replace("$", ""))
df["price_paid"] = df["price_paid"].astype(float)

df["price_paid"].mean()
```

## Somme avec `sum()`

On peut obtenir le **total** d'une colonne numérique avec **`sum()`**.

```python
df["price_paid"].sum()
```

## Minimum et Maximum

On retrouve la **valeur minimum** et la **valeur maximum** d'une colonne avec **`min()`** et **`max()`**.

```python
df["price_paid"].min()
```

```python
df["price_paid"].max()
```

## Valeurs uniques avec `unique()`

Pour récupérer les **valeurs uniques** d'une colonne, on utilise **`unique()`**.

```python
df["country"].unique()
```

### Conversion en liste avec `tolist()`

On peut convertir le résultat en **liste Python** avec **`tolist()`**.

```python
df["price_paid"].unique().tolist()
```

## Compter les occurrences avec `value_counts()`

**`value_counts()`** renvoie le **nombre d'occurrences** de chaque valeur dans une colonne. On peut aussi passer `normalize=True` pour obtenir les **proportions**.

```python
df["country"].value_counts()
```

```python
df["gender"].value_counts()

# ou 

df["gender"].value_counts(normalize=True)
```

Pour **regrouper** des éléments selon une colonne, on utilise **`groupby()`**.

```python
# ne peut pas etre utilisé comme ceci 
x =list(df.groupby("country"))
x
```
