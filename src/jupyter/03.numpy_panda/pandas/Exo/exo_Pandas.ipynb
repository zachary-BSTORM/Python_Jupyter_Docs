### Exercice 1
Créer une **Series** à partir de la liste `[2, 4, 6, 8, 10]` et l'afficher.

### Exercice 2
Additionner, soustraire, multiplier et diviser les deux Series suivantes :

`[2, 4, 6, 8, 10]` et `[1, 3, 5, 7, 9]`

```python
# Créer les Series
serie1 = pd.Series([2, 4, 6, 8, 10])
serie2 = pd.Series([1, 3, 5, 7, 9])

# Addition
addition = serie1 + serie2

# Soustraction
soustraction = serie1 - serie2

# Multiplication
multiplication = serie1 * serie2

# Division
division = serie1 / serie2

# Afficher les résultats
print("Addition:\n", addition)
print("Soustraction:\n", soustraction)
print("Multiplication:\n", multiplication)
print("Division:\n", division)
```

### Exercice 3

### Exercice 4
Ajouter une série contenant des objets à une autre série.

```python
# Créer les Series
serie1 = pd.Series(['100', '200', 'python', '300.12', '400'])
serie2 = pd.Series(['500', 'php'])

# Ajouter la serie2 à la serie1
somme = pd.concat([serie1, serie2], ignore_index=True)

# Afficher le résultat
print(somme)
```

### Exercice 8
Extraire les lignes où les données sont manquantes (**NaN**).

```python
import pandas as pd
import numpy as np

# Dictionnaire de données
exam_data = {
    'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily', 'Michael', 'Matthew', 'Laura', 'Kevin', 'Jonas'],
    'score': [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],
    'attempts': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    'qualify': ['yes', 'no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'no', 'yes']
}

# Labels pour l'index
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']

# Créer le DataFrame
df = pd.DataFrame(exam_data, index=labels)

# Extraire les lignes où les données sont manquantes (NaN)
df_nan = df[df.isna().any(axis=1)]

# Afficher les lignes avec des NaN
print(df_nan)
```

### Exercice 9
Extraire les lignes où `attempts < 2` **et** `score > 15`.

```python
import pandas as pd
import numpy as np

# Dictionnaire de données
exam_data = {
    'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily', 'Michael', 'Matthew', 'Laura', 'Kevin', 'Jonas'],
    'score': [ 12.5, 9 , 16.5, np.nan, 9, 20, 14.5, np.nan, 8 , 19],
    'attempts': [1 , 3 , 2   , 3     , 2, 3 , 1   , 1     , 2 , 1 ],
    'qualify': ['yes', 'no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'no', 'yes']
}

# Labels pour l'index
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']

# Créer le DataFrame
df = pd.DataFrame(exam_data, index=labels)

# Extraire les lignes où 'attempts' < 2 et 'score' > 15
df_filtered = df[(df['attempts'] < 2) & (df['score'] > 15)]

# Afficher les lignes filtrées
print(df_filtered)
```

### Exercice 10

### Exercice 11
Remplacer `'yes'` et `'no'` dans la colonne `qualify` par des **booléens**, puis ajouter une nouvelle ligne.

```python
import pandas as pd
import numpy as np

# Dictionnaire de données
exam_data = {
    'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily', 'Michael', 'Matthew', 'Laura', 'Kevin', 'Jonas'],
    'score': [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],
    'attempts': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    'qualify': ['yes', 'no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'no', 'yes']
}

# Labels pour l'index
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']

# Créer le DataFrame
df = pd.DataFrame(exam_data, index=labels)

# Remplacer 'yes' par True et 'no' par False dans la colonne 'qualify'
df['qualify'] = df['qualify'].replace({'yes': True, 'no': False})

# Ajouter la nouvelle ligne avec les valeurs spécifiées
df.loc['k'] = ['Suresh', 15.5, 1, True]

# Afficher le DataFrame mis à jour
print(df)
```
