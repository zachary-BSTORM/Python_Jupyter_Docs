# Structure de données avec Pandas

**Series** : tableau 1D indexé
- Une **Series** dans Pandas est un tableau 1D étiqueté
- C'est similaire à un tableau NumPy, mais avec des étiquettes d'index personnalisables
- Elle peut contenir n'importe quel type de données : nombres, chaînes, booléens, objets, etc.

**DataFrame** : tableau 2D indexé (lignes et colonnes), similaire à une feuille Excel ou une table SQL.

````python
import pandas as pd

data = pd.Series([0.25, 0.5, 0.75, 1.0])

data.index    # IndexRange object ( une formule compressée ) : range(0, 4)
list(data.index)  # Tableau numpy des indices : array([0, 1, 2, 3])
data.values   # Tableau numpy des valeurs : array([0.25, 0.5, 0.75, 1.0])
````

## Index explicites (noms personnalisés)

Contrairement aux tableaux NumPy, il est possible de donner des noms personnalisés aux index :

````python
data = pd.Series([0.25, 0.5, 0.75, 1.0], index=['a', 'b', 'c', 'd'])

data = pd.Series([0.25, 0.5, 0.75, 1.0], index=[2, 5, 3, 7])
````

## DataFrame

Un **DataFrame** ressemble à un array 2D mais avec plus de flexibilité. On peut par exemple ajouter des noms aux colonnes. Un DataFrame est au minimum constitué de deux **Series** (deux colonnes).

- Chaque colonne est une **Series** avec un nom
- Chaque ligne est identifiée par un **index** (comme dans les Series)
- Il peut contenir plusieurs types de données (nombres, chaînes, booléens…)

| Propriété/Méthode | Description |
|---|---|
| `df.index` | Index des lignes (par défaut : 0, 1, 2…) |
| `df.columns` | Liste des noms de colonnes |
| `df.values` | Données sous forme de tableau NumPy |
| `df.dtypes` | Type de chaque colonne |
| `df.shape` | Dimensions du tableau (lignes, colonnes) |
| `df.mean(numeric_only=True)` | Moyenne par colonne |
| `df.sum(numeric_only=True)` | Somme par colonne |
| `df.sort_values("col")` | Trier par colonne |
| `df.groupby("col").mean()` | Grouper par colonne |
| `df.drop("col", axis=1)` | Supprimer une colonne |

````python
data = {
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "city": ["Paris", "Lyon", "Marseille"]
}

df = pd.DataFrame(data)
````

````python
df["name"]           # Retourne une Series (une seule colonne)

df[["name", "age"]]  # Retourne un DataFrame (plusieurs colonnes)

df.loc['e']         # Par étiquette (index) → accès à la ligne 0 ( loc contrairement à iloc permet d'accéder à une ligne par son étiquette )

df.iloc[1]        # Par position → accès à la 2e ligne ( iloc permet d'accéder à une ligne par sa position )

df[df["age"] > 28]   # Sélectionne les personnes de plus de 28 ans
````

## Indexation et sélection dans un DataFrame

````python
data = pd.DataFrame(
{
    'age': [22, 35, 58, 44],
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'score': [88.5, 92.3, 70.0, 85.2]
}, 
index=['a', 'b', 'c', 'd'])

data
````

## Opérations et alignement des index

Les opérations entre **Series** ou **DataFrame** s'alignent automatiquement sur les index.

- Si un index est manquant dans l'un des deux objets, le résultat sera **NaN** (Not a Number)
- On peut fournir une valeur par défaut avec `fill_value` quand un index est absent
- Les fonctions NumPy (comme `np.sqrt`, `np.exp`, etc.) fonctionnent directement sur les objets Pandas

### Opération avec des indices décalés

````python
A = pd.Series([2, 4, 6], index=[0, 1, 2])
B = pd.Series([1, 3, 5], index=[1, 2, 3])

A + B
````

### Éviter les NaN avec `.add()` et `fill_value`

````python
A.add(B, fill_value=0)
````

### Utilisation de fonctions NumPy

````python
np.sqrt(A)
````

### Opérations sur DataFrame

````python
df1 = pd.DataFrame({
    "a": [1, 2, 3],
    "b": [4, 5, 6]
})

df2 = pd.DataFrame({
    "a": [7, 8, 9],
    "b": [10, 11, 12]
})

df1 + df2
````

### fill_value sur DataFrames

Définir les valeurs manquantes à `0` lors d'opérations entre DataFrames.

````python
df1.add(df2, fill_value=0)
````

## Combiner des Datasets

### Concaténation verticale (`axis=0`)

Par défaut, `pd.concat` concatène verticalement (empile les lignes).

````python
ser1 = pd.Series(['A', 'B', 'C'], index=[1, 2, 3])
ser2 = pd.Series(['D', 'E', 'F'], index=[4, 5, 6])

print(pd.concat([ser1, ser2]))
````

### Concaténation horizontale (`axis=1`)

Forme un DataFrame en alignant les index.

````python
print(pd.concat([ser1, ser2], axis=1))
````

### Vérifier l'intégrité des index

L'option `verify_integrity=True` force une vérification des doublons dans les index. En cas de doublon, une erreur est levée.

````python

x = pd.Series(['A', 'B'], index=[0, 1])
y = pd.Series(['C', 'D'], index=[3, 2])

# Lève une erreur car l'index 1 est dupliqué
pd.concat([x, y], verify_integrity=True)
````

### Ignorer les index avec `ignore_index`

Ignore les index d'origine et en crée de nouveaux numériques de 0 à n.

````python
pd.concat([ser1, ser2], ignore_index=True)
````

### Garder les colonnes communes (`join='inner'`)

Concatène en ne gardant que les colonnes communes. Par défaut `join='outer'` (toutes les colonnes).

````python
df5 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df6 = pd.DataFrame({'B': [5, 6], 'C': [7, 8]})

pd.concat([df5, df6], join='inner')
````

### Filtrer les colonnes manuellement

> `join_axes` est **déprécié** dans les versions récentes de Pandas. On filtre les colonnes manuellement.

````python
df5 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df6 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})

pd.concat([df5, df6[df5.columns]], axis=0)
````

## Fusion de DataFrames (merge)

> Équivalent du **JOIN** en SQL.

```python
pd.merge(df1, df2, how='inner', on='colonne_commune')
```

| Paramètre | Description |
|---|---|
| `left`, `right` | Les deux DataFrames à fusionner |
| `how` | Type de jointure : `'inner'`, `'left'`, `'right'`, `'outer'` |
| `on` | Colonne(s) commune(s) à utiliser pour la jointure |
| `left_on` | Colonne(s) du DataFrame de gauche à utiliser |
| `right_on` | Colonne(s) du DataFrame de droite à utiliser |
| `left_index` | Utilise l'index du DataFrame de gauche |
| `right_index` | Utilise l'index du DataFrame de droite |
| `suffixes` | Suffixes à ajouter pour les colonnes dupliquées (ex. `('_x', '_y')`) |
| `indicator=True` | Ajoute une colonne `_merge` pour indiquer l'origine des lignes |

### Jointure simple avec colonne commune

````python
# DataFrame 1 avec un id
df1 = pd.DataFrame({
    'id': [1, 2, 3],
    'A': ['A1', 'A2', 'A3'],
    'B': [10, 20, 30]
})

# DataFrame 2 avec un id
df2 = pd.DataFrame({
    'id': [3, 4, 5],
    'C': ['C3', 'C4', 'C5'],
    'D': [300, 400, 500]
})

pd.merge(df1, df2, on="id")
````

### Types de jointures (`how`)

| Type | Description |
|---|---|
| `inner` | Garde les lignes avec des clés communes |
| `left` | Garde toutes les lignes de `df1`, et celles de `df2` si correspondance |
| `right` | Garde toutes les lignes de `df2`, et celles de `df1` si correspondance |
| `outer` | Garde tout (union), remplit les NaN si pas de correspondance |

````python
pd.merge(df1, df2, on="id", how="outer")
````

### Colonnes avec des noms différents

````python
pd.merge(df1, df2, left_on="id", right_on="id")
````

### Fusionner sur les index

````python
# signifie que la fusion doit être effectuée en utilisant les index des deux DataFrames
pd.merge(df1, df2, left_index=True, right_index=True)
````

### Ajouter un indicateur

````python
pd.merge(df1, df2, on="id", how="outer", indicator=True)
````

### Gérer les colonnes dupliquées

````python
pd.merge(df1, df2, on="id", suffixes=('_gauche', '_droite'))
````

## Analyse statistique avec `describe()`

`describe()` fournit un résumé statistique des colonnes numériques.

````python
df = pd.DataFrame({
    'age': [22, 35, 58, 41],
    'taille': [1.65, 1.80, 1.75, 1.60],
    'nom': ['Alice', 'Bob', 'Charlie', 'David']
})

df.describe()
````

| Paramètre | Description |
|---|---|
| `include='all'` | Inclut toutes les colonnes (numériques, objets, booléens, etc.) |
| `include=['object']` | Décrit uniquement les colonnes de type `object` (texte, catégories, etc.) |
| `exclude=['number']` | Exclut les colonnes numériques |
| `percentiles=[...]` | Personnalise les centiles à afficher (par défaut : 25%, 50%, 75%) |

````python
df.describe(include='all')
df.describe(include=['object'])
df.describe(percentiles=[.1, .5, .9])
````

## Autres méthodes statistiques

| Méthode | Description |
|---|---|
| `df.mean()` | Moyenne par colonne |
| `df.median()` | Médiane par colonne |
| `df.std()` | Écart-type |
| `df.var()` | Variance |

## L'agrégation

Résumer des données à l'aide de fonctions comme `sum()`, `mean()`, `count()`, etc. L'agrégation est souvent utilisée avec `groupby()` pour obtenir des statistiques par groupe.

````python
df["colonne"].sum()
df["colonne"].mean()
df["colonne"].min()
df["colonne"].max()
df["colonne"].count()
df["colonne"].nunique()   # nombre de valeurs uniques
````

## GroupBy

`groupby()` permet de regrouper les données selon une ou plusieurs colonnes pour effectuer des calculs par groupe.

````python
df = pd.DataFrame({
    'genre': ['H', 'F', 'F', 'H', 'F'],
    'age': [23, 45, 36, 52, 33],
    'revenu': [1000, 1200, 1300, 1100, 1250]
})

# Moyenne de l'âge et du revenu par genre
df.groupby("genre")[["age", "revenu"]].mean()
````

### `agg()` pour plusieurs agrégations

| Méthode | Description |
|---|---|
| `groupby()` | Regroupe les lignes selon une ou plusieurs colonnes |
| `agg()` | Applique une ou plusieurs fonctions sur chaque groupe |
| `reset_index()` | Remet l'index à plat après un `groupby()` |
| `as_index=False` | Évite que la colonne de regroupement devienne l'index |

````python
df[ "revenu" ] .agg ( ["min", "max", "mean"] )
````

````python
# Exemple de DataFrame
df = pd.DataFrame({
    'genre': ['Homme', 'Femme', 'Homme', 'Femme', 'Homme'],
    'age': [25, 30, 35, 40, 45],
    'revenu': [2000, 3000, 2500, 4000, 3500]
})

# GroupBy avec agrégation multiple
result = df.groupby("genre", as_index=False).agg(
    {
    "age": "mean",  # Moyenne des âges
    "revenu": ["min", "max", "mean"]  # Min, Max, et Moyenne des revenus
    }
    )

print(result)
````
