# Sélectionner des données

En utilisant les **crochets** `[]`, on peut préciser le nom de la colonne que l'on souhaite récupérer :

```python
df["email"]
```

- On peut chaîner la méthode **`head()`** pour n'afficher que les premières valeurs.
- Si le nom de la colonne contient des **espaces**, cette notation est **obligatoire**.

````python
import pandas as pd

df = pd.read_csv("../data.csv")

df["email"].head()
````

## Accès par attribut (notation pointée)

On peut également récupérer une colonne avec la **notation pointée**, à condition que le nom de la colonne **ne contienne pas d'espaces** :

```python
df.email
```

La méthode **`head()`** fonctionne aussi avec cette syntaxe.

````python
df.email
df.email.head()
````

## DataFrame vs Series

Pandas manipule deux types d'objets :

| Type | Description |
|------|-------------|
| **DataFrame** | Tableau complet à **2 dimensions** (lignes + colonnes) |
| **Series** | Une seule colonne du DataFrame, structure **1D** |

> Cette distinction est importante : certaines opérations ne sont disponibles que sur l'un ou l'autre type.

On peut utiliser la fonction **`type()`** pour vérifier le type d'un objet :

```python
type(df)        # → DataFrame
type(df.email)  # → Series
```

````python
#type(df)

type(df.email)
````

### Opérations spécifiques aux Series

| Opération | Description |
|-----------|-------------|
| `str.` | Méthodes spécifiques aux chaînes de caractères (ex : `str.lower()`) |
| `dt.` | Méthodes spécifiques aux dates (ex : `dt.year`, `dt.month`) |
| `map()` | Appliquer une fonction **valeur par valeur** |
| `Series[index]` | Accès direct à une valeur (pas besoin de `.loc[]`) |

### Opérations spécifiques aux DataFrames

| Opération | Description |
|-----------|-------------|
| `df['col']` ou `df[['col1', 'col2']]` | Sélection de colonnes |
| `df.loc[row, col]`, `df.iloc[...]` | Sélections avancées ligne + colonne |
| `groupby()` | Regroupement par colonnes |
| `pivot_table()` | Création de tableaux croisés dynamiques |
| `merge()`, `join()` | Fusionner plusieurs DataFrames |
| `drop(columns=...)` | Supprimer des colonnes entières |
| `corr()`, `cov()` | Corrélation / covariance entre colonnes numériques |
| `duplicated()` / `drop_duplicates()` | Identifier / retirer les doublons sur plusieurs colonnes |
| `set_index()` / `reset_index()` | Modifier ou réinitialiser l'index |

### Opérations communes

Ces opérations fonctionnent à la fois sur les **Series** et les **DataFrames** :

| Opération | Description |
|-----------|-------------|
| `head(n)` / `tail(n)` | Affiche les premières / dernières lignes |
| `shape`, `size`, `ndim` | Dimensions, taille, nombre d'axes |
| `describe()` | Statistiques de base (moyenne, min, max, etc.) |
| `value_counts()` | Compter les occurrences (surtout utile en Series) |
| `astype()` | Changer le type de données |
| `isnull()`, `notnull()` | Détection de valeurs manquantes |
| `apply()` | Appliquer une fonction (à chaque ligne/élément) |
| `sort_values()` | Trier par valeur |
| `unique()`, `nunique()` | Valeurs uniques et nombre de valeurs uniques |
| `mean()`, `sum()`, etc. | Calculs numériques simples |
| `plot()` | Graphiques (via matplotlib ou seaborn) |

## Slicing avec les crochets

On peut utiliser les **crochets** pour récupérer un **intervalle de lignes**, comme un **slice** sur les listes en Python :

```python
df[10:20]  # lignes d'index 10 à 19 (borne max exclue)
```

````python
df[10:20]
````

## Sélection avec `loc[]`

La méthode **`loc[]`** permet de sélectionner des lignes par leur **index** :

```python
df.loc[10:13]
```

> **Différence clé** : contrairement au slicing classique, la **borne maximale est incluse** avec `loc[]`.

````python
df.loc[10:13]
````

## Changer l'index et utiliser `loc[]`

Avec **`set_index()`**, on peut remplacer l'index numérique par une colonne (ex : `"email"`). On utilise ensuite **`loc[]`** pour cibler des lignes par cette nouvelle clé :

```python
df_email = df.set_index("email")
df_email.loc["lwoodcrafta@sitemeter.com"]
```

> Lorsqu'on cible un **élément précis**, `loc[]` renvoie un objet **Series**.

Pour convertir le résultat en objet Python natif, on peut utiliser **`.values.tolist()`**.

````python
df_email = df.set_index("email")

df_email.loc["lwoodcrafta@sitemeter.com"]

# on a la possibilité de passer plusieurs informations en paramètre
# df_email.loc["lwoodcrafta@sitemeter.com","lwoodcrafta@sitemeter.com","lwoodcrafta@sitemeter.com"]
````
