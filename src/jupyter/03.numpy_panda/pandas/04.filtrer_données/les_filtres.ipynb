# Filtrer les données

```python
import pandas as pd

df = pd.read_csv("../data.csv")

# dans le cas d'une condition simple on va retrouver des booleans en guise de reponse
df["gender"] == "Male"
```

## Appliquer un filtre

Pour récupérer les éléments qui correspondent à une condition, on passe la **condition** directement à l'intérieur des **crochets** `[]` du DataFrame.

On obtient ainsi uniquement les lignes où la condition est `True`.

```python
df[df["gender"] == "Male"]

# on peut enregistrer le filtre 

female_filter = df["gender"] == "Female"

df[female_filter]
```

## Filtrer avec `isin()`

Pour vérifier si des valeurs font partie d'une liste donnée, on utilise la méthode **`isin()`**.

```python
# on peut passer une liste ou un tuple
# on va retrouver des booleans en guise de reponse
# ne pas hesiter a regarder la documentation de pandas en fonction de ce que vous voulez faire

# https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.isin.html

# ici une liste
df["country"].isin(("France", "Germany"))

# ici un tuple
df["country"].isin(["France", "Germany"])

country_filter = df["country"].isin(["France", "Germany"])

df[country_filter]
```

## Attention aux types de données

La colonne **`price_paid`** est de type `str` (ex: `"$5.36"`). On ne peut donc pas directement appliquer un filtre numérique dessus.

```python
df[df["price_paid"] > 6]

# TypeError: '>' not supported between instances of 'str' and 'int'
```

### Conversion de type avant filtrage

On peut convertir la colonne en supprimant le caractère `$` puis en changeant le type en `float`.

```python
df_test = df.copy()
df_test.price_paid = df_test.price_paid.apply(lambda x: x.replace("$", ""))
df_test.price_paid = df_test.price_paid.astype(float)
df_test[df_test["price_paid"] > 6]
```
