RÉCAP DES RÈGLES AVEC NUMPY

1 : Tous les éléments d’un ndarray ont le même type

2 : Le slicing crée une vue (view), pas une copie

3 : Les formes (shapes) doivent être compatibles pour opérations

4 : Différence entre shape, ndim, size, dtype

shape	Dimensions du tableau (lignes, colonnes, ...)
ndim	Nombre de dimensions
size	Nombre total d’éléments
dtype	Type de données

5 : Broadcasting NumPy étend automatiquement les dimensions pour permettre les opérations.

6 : Mémoire contiguë et performance NumPy est optimisé : les opérations vectorisées sont plus rapides que des boucles.

7 : Les dimensions doivent être compatibles pour le reshape

8 : Indexation logique (booléenne) : 
- print(a[a > 25])

9 : Opérations élément par élément

10 : Transposition ≠ reshape
.T inverse lignes ↔ colonnes, mais ne change pas la structure mémoire.



np.array()

Créer un tableau NumPy

```python
import numpy as np
a = np.array([1, 2, 3])
print(a)  # [1 2 3]
```

np.arange(start, stop, step)

Créer un tableau d'entiers espacés régulièrement (comme range)

```python
a = np.arange(1, 10, 2)
print(a)  # [1 3 5 7 9]
```

np.linspace(start, stop, num)

Créer un tableau de num valeurs espacées régulièrement entre start et stop

```python
a = np.linspace(0, 1, 5)
print(a)  # [0.   0.25 0.5  0.75 1.  ]
```

np.zeros(shape) / np.ones(shape)

Créer un tableau rempli de 0 ou 1

```python
print(np.zeros((2, 3)))  # 2 lignes, 3 colonnes de zéros
print(np.ones((2, 2)))   # 2 lignes, 2 colonnes de uns
```

np.eye(n)

Créer une matrice identité (diagonale à 1)

```python
print(np.eye(3))
# [[1. 0. 0.]
#  [0. 1. 0.]
#  [0. 0. 1.]]
```

ndarray.T

Transposer (inverser lignes et colonnes)

```python
a = np.array([[1, 2], [3, 4]])
print(a.T)
# [[1 3]
#  [2 4]]
```

np.concatenate((a, b), axis=0 or 1)

Coller deux tableaux ensemble (en ligne ou colonne)

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6]])
print(np.concatenate((a, b), axis=0))  # Ajout en ligne
```

np.hstack((a, b)) / np.vstack((a, b))

Empiler horizontalement ou verticalement

```python
a = np.array([1, 2])
b = np.array([3, 4])
print(np.hstack((a, b)))  # [1 2 3 4]
print(np.vstack((a, b)))  # [[1 2]
                          #  [3 4]]
```

x.T — Transposition en NumPy


Il retourne la transposée du tableau.

Il inverse les axes :

les lignes deviennent des colonnes

les colonnes deviennent des lignes

Pour un tableau 2D (m x n), la transposition donne un tableau n x m.

ne s'effectue pas sur les tableaux 1D

inverse tout les axes sur les tableaux 3D

```python
# x.T


x = np.array([[1, 2, 3],
              [4, 5, 6]])

print("Original :\n", x)
print("Transposé :\n", x.T)
```

Transformer vecteur en matrice colonne/ligne

 Objectif : transformer un vecteur 1D en matrice ligne ou matrice colonne

 Un vecteur 1D : c’est juste une ligne d’éléments à plat.

```python
x = np.array([1, 2, 3])
print(x.shape)  # (3,) → une seule ligne, pas de colonne

# Mais pour faire des matrices (tableaux 2D), il faut dire explicitement :

# Combien de lignes ?

# Combien de colonnes ?
```

Transformer x = [1, 2, 3] en matrice ligne

```python
# Pour obtenir ceci : 
[[1, 2, 3]]   # 1 ligne, 3 colonnes
```

Solution 1 : reshape((1, 3))
je veux 1 ligne, 3 colonnes

```python
x.reshape((1, 3))

# [[1 2 3]]
```

Solution 2 : x[np.newaxis, :]
ajoute une nouvelle ligne

```python
x[np.newaxis, :]  # On garde toutes les colonnes (:) et on ajoute une ligne

# [[1 2 3]]
```

Transformer x = [1, 2, 3] en matrice colonne

Pour obtenir ceci : 

[[1],
 [2],
 [3]]

 Solution 1 : reshape((3, 1))
je veux 3 lignes, 1 colonne

```python
x.reshape((3, 1))

# [[1]
#  [2]
#  [3]]
```

 Solution 2 : x[:, np.newaxis]
je garde toutes les lignes (:) et j’ajoute une nouvelle colonne

```python
x[:, np.newaxis]

# [[1]
#  [2]
#  [3]]
```

Vecteur ligne	x.reshape(1, 3)	[[1, 2, 3]]	(1, 3)

x[np.newaxis, :]	[[1, 2, 3]]	(1, 3)

Vecteur colonne	x.reshape(3, 1)	[[1], [2], [3]]	(3, 1)

x[:, np.newaxis]	[[1], [2], [3]]	(3, 1)

np.concatenate

np.concatenate permet de coller plusieurs tableaux ensemble 

Le paramètre axis indique comment on les colle :

- axis=0 → on colle par en dessous (ligne après ligne)

- axis=1 → on colle par le côté (colonne après colonne)

Les dimensions doivent être compatibles pour l'axe non utilisé.

```python
np.concatenate([x, x], axis=0)
```
