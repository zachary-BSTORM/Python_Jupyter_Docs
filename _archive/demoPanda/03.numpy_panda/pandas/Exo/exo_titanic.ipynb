```python
import pandas as pd
import matplotlib.pyplot as plt

#pip install xlrd


# import du fichier excel du titanic
df = pd.read_excel("../titanic3.xls")

```

```python
# Affichage des 5 premières lignes
df.head()

# Affichage des 5 dernières lignes
df.tail()

# Affichage des informations sur le dataframe
df.info()

# Affichage des statistiques descriptives
df.describe()

# Affichage des colonnes
df.columns

# Affichage des valeurs uniques de la colonne 'pclass'
df['pclass'].unique()

# Affichage des valeurs uniques de la colonne 'name'
df['name'].unique()

# statistiques rapides
df.describe()

# eliminer certaines colonnes
df.drop(['ticket'])

# eliminer les lignes avec des valeurs manquantes
df.dropna(subset=['age'])

# compter les repetitions
df['pclass'].value_counts()

# analyse par groupe
df.groupby(['pclass'])
```
