Additionner, soustraire, multiplier et diviser les deux Series suivantes :
Sample Series : [2, 4, 6, 8, 10], [1, 3, 5, 7, 9]

Effectuer les opérations d'addition, soustraction, multiplication et division entre ces deux séries

```python
# Créer les Series
serie1 = pd.Series([2, 4, 6, 8, 10])
serie2 = pd.Series([1, 3, 5, 7, 9])

# Addition
addition = serie1 + serie2

# Soustraction
soustraction = serie1 - serie2

# Multiplication
multiplication = serie1 * serie2

# Division
division = serie1 / serie2

# Afficher les résultats
print("Addition:\n", addition)
print("Soustraction:\n", soustraction)
print("Multiplication:\n", multiplication)
print("Division:\n", division)
```

11. Remplacer les valeurs ‘yes’ et ‘no’ dans la colonne ‘qualify’ par un
Booléen.
Sample DataFrame:
exam_data = {'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily',
'Michael', 'Matthew', 'Laura', 'Kevin', 'Jonas'],
'score': [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],
'attempts': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
'qualify': ['yes', 'no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'no', 'yes']}
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']
Values for each column will be:
name : ‘Suresh’, score: 15.5, attempts: 1, qualify: ‘yes’, label: ‘k’

```python
import pandas as pd
import numpy as np

# Dictionnaire de données
exam_data = {
    'name': ['Anastasia', 'Dima', 'Katherine', 'James', 'Emily', 'Michael', 'Matthew', 'Laura', 'Kevin', 'Jonas'],
    'score': [12.5, 9, 16.5, np.nan, 9, 20, 14.5, np.nan, 8, 19],
    'attempts': [1, 3, 2, 3, 2, 3, 1, 1, 2, 1],
    'qualify': ['yes', 'no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'no', 'yes']
}

# Labels pour l'index
labels = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j']

# Créer le DataFrame
df = pd.DataFrame(exam_data, index=labels)

# Remplacer 'yes' par True et 'no' par False dans la colonne 'qualify'
df['qualify'] = df['qualify'].replace({'yes': True, 'no': False})

# Ajouter la nouvelle ligne avec les valeurs spécifiées
df.loc['k'] = ['Suresh', 15.5, 1, True]

# Afficher le DataFrame mis à jour
print(df)
```
