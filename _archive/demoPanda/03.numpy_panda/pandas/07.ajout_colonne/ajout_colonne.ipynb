Nous avons la possibilitées d'ajouter des colonnes

pour par exemple ajouter un total sur nos rangées

donc ici on va commencer par modifier les valeurs null 

puis effectuer une operation pour ajouter et definir le total


```python
import pandas as pd

# Chargement du fichier CSV
df = pd.read_csv("../data.csv")

# Remplissage des valeurs manquantes dans la colonne "tax" par 0
df["tax"] = df["tax"].fillna(0)  # Important d’assigner le résultat à la colonne

# Suppression du symbole "$" dans la colonne "price_paid" puis conversion en float
df["price_paid"] = df["price_paid"].apply(lambda x: x.replace("$", ""))
df["price_paid"] = df["price_paid"].astype(float)

# Vérification des types des colonnes
df.dtypes

```

donc si nous desirons ajouter une colonne nous utiliserons les crochets avec le nom de la nouvelle colonne

- df["test"] = 0 : pour ajouter une colonne

- del df["test"] : pour supprimer une colonne

donc maintenant nous allons creer une colonne :  prix_total

```python
# Remplir les valeurs manquantes de la colonne "tax" avec 0
df["tax"] = df["tax"].fillna(0) # Important d’assigner le résultat à la colonne

# Calcul du prix total après application de la taxe
df["total_price"] = df["price_paid"] * (1 - df["tax"] / 100)


df
```

Imaginons que l'on veuille ajouter un code pour les pays ici on va parler de mapping avec la methode 

- map()

on va d'abord creer un dictionnaire

```python
countries = {"United States": "USA", "United Kingdom": "UK", "France": "FR", "Germany": "DE" , "Morocco": "MA" , "Canada" : "CA"}
# Remplacement des noms de pays par leurs abréviations

df["country_code"] = df["country"].map(countries)

df.head()
```
