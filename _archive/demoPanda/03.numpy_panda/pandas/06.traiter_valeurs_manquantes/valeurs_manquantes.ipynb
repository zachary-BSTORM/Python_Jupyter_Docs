pour verifier si quelque chose est null nous aurons plusieurs possibilitées

- df.isnull()  : va nous renvoyer un booleen si la case est null

- df.notnull() : va nous renvoyer true si la case n'est pas null

et nous allons pouvoir l'utiliser dans nos filtres pour recuperer les elements non null ou null

```python
import pandas as pd

df = pd.read_csv("../data.csv")

df.isnull()
```

Pour l'utiliser dans un filtre

```python
#df[df["tax"].isnull()]

df[df["tax"].notnull()]
```

Ensuite pour modifier les valeurs null on retrouve la methode 

- fillna()

qui peut prendre plusieurs paramètres

- 0 : qui remplace les valeurs null ou nan 

- method='bfill' : qui va utiliser les valeurs autour pour remplir les cases ( pas forcement une bonne pratique vu que l'on modifie les données )

```python
df["tax"].fillna(0)

df["tax"].fillna(method='bfill')
```

Ensuite la methode 

- dropna() : permet d'enlever les paramètres null 

sans paramètre retire toutes les rangée qui contiennent null 

on peut cibler une ou plusieurs colonnes

- dropna(subset=["tax"]) : permet de retirer une ou plusieurs colonnes

dispose egalement de : inplace=True

```python
# df_copy = df.dropna()

# df_copy = df.dropna(subset=["tax"],inplace=True)

df_copy = df.dropna(subset=["tax"])
```
