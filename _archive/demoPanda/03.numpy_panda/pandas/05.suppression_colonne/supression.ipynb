Nous allons pouvoir supprimer une colonne avec la methode 

- drop()

- axis = 0  # on va supprimer les rangées

- axis = 1 # on va supprimer les colonnes

```python
import pandas as pd

df = pd.read_csv("../data.csv")

df_test = df.drop("ip_address", axis=1)

# df.drop("ip_address", axis=1 , inplace=True) # permet de travailler sur la source  



df_test
```

drop() peut prendre plusieurs paramètres

```python
df_copy = df.drop(["first_name","last_name","email"],axis=1)

df_copy
```
