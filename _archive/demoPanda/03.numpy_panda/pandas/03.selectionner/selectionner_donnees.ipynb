deux types d'objet :

- dataframe : tableau complet ( 2D )

- series    : une seule colonne de notre df ( 1D )

cette informations est importante car nous ne pourrons effectuer certaines actions que sur l'un ou l'autre type 

on peut utiliser la methode type() pour recuperer le type de notre liste ou de notre colonne

- type()

```python
type(df)

type(df.email)
```

operations uniquement disponible sur 

- Serie

str. ------------Méthodes spécifiques aux chaînes de caractères (ex : str.lower())

dt.  ------------Méthodes spécifiques aux dates (ex : dt.year, dt.month)

map() -----------Appliquer une fonction valeur par valeur

Series[index] ---Accès direct à une valeur (pas besoin de .loc[])

operations uniquement disponible sur 

- DataFrame

df['colonne'] ou df[['col1', 'col2']]  ----Sélection de colonnes
df.loc[row, col], df.iloc[...]	-----------Sélections avancées ligne + colonne
groupby()   -------------------------------Regroupement par colonnes
pivot_table()  ----------------------------Création de tableaux croisés dynamiques
merge(), join()  --------------------------Fusionner plusieurs DataFrames
drop(columns=...)  ------------------------Supprimer des colonnes entières
corr(), cov()  ----------------------------Corrélation / covariance entre colonnes numériques
duplicated() / drop_duplicates()  ---------Identifier / retirer les doublons sur plusieurs colonnes
set_index() / reset_index()  --------------Modifier ou réinitialiser l’index

operations disponible sur les deux 

- Serie

- DataFrame

head(n) / tail(n)  ----------Affiche les premières/dernières lignes
shape, size, ndim  ----------Dimensions, taille, nombre d'axes
describe()  -----------------Statistiques de base (moyenne, min, max, etc.)
value_counts()  -------------Compter les occurrences (surtout utile en Series)
astype()  -------------------Changer le type de données
isnull(), notnull()  --------Détection de valeurs manquantes
apply()  --------------------Appliquer une fonction (à chaque ligne/élément)
sort_values()  --------------Trier (par une colonne pour DF, par valeur pour Series)
unique(), nunique()  --------Valeurs uniques et nombre de valeurs uniques
mean(), sum(), etc.  --------Moyenne, somme, etc. (calculs numériques simples)
plot()  ---------------------Graphiques (via matplotlib ou seaborn)

on peut egalement utiliser les crochets pour recuperer un interval 

- [10:20]

comme un slice sur les listes en python

```python
df[10:20]
```

on peut utiliser la methode 

- loc[10:13]

la difference est que la borne max est inclus 

permet de cibler l'index 

```python
df.loc[10:13]
```

avec la mathode loc[] nous allons pouvoir  cibler un autre type en changeant l'index 

ici on recupere un objet quand on demande un element precis

on peut aussi utiliser le 

- .values.tolist() 

pour recuperer un objet python

```python
df_email = df.set_index("email")

df_email.loc["lwoodcrafta@sitemeter.com"]

# on a la possibilité de passer plusieurs informations en paramètre
# df_email.loc["lwoodcrafta@sitemeter.com","lwoodcrafta@sitemeter.com","lwoodcrafta@sitemeter.com"]
```
