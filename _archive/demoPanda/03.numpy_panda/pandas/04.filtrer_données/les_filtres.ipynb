Nous allons pouvoir filtrer les données 



```python
import pandas as pd

df = pd.read_csv("../data.csv")

# dans le cas d'une condition simple on va retrouver des booleans en guise de reponse
df["gender"] == "Male"
```

pour recuperer les element qui correspondent a notre condition on va avoir besoin de passer toute la condition dans les crochets

il faut bien comprendre qu'on a une condition a l'intérieur des crochets et notre objet qui selectionne quelque-chose

```python
df[df["gender"] == "Male"]

# on peut enregistrer le filtre 

female_filter = df["gender"] == "Female"

df[female_filter]
```

pour effectuer certaines actions nous allons devoir utiliser certaines methodes disponible comme

- isin()

```python
# on peut passer une liste ou un tuple
# on va retrouver des booleans en guise de reponse
# ne pas hesiter a regarder la documentation de pandas en fonction de ce que vous voulez faire

# https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.isin.html

# ici une liste
df["country"].isin(("France", "Germany"))

# ici un tuple
df["country"].isin(["France", "Germany"])

country_filter = df["country"].isin(["France", "Germany"])

df[country_filter]
```

parfois il y a quelques operations a faire , par exemple ici le prix est en chaine de caractères ( $5.36)

```python
df[df["price_paid"] > 6]

# TypeError: '>' not supported between instances of 'str' and 'int'
```

on pourrait effectuer une modification sur le champ concerné mais c'est plus avancé voici l'exemple pour les curieux

```python
df_test = df.copy()
df_test.price_paid = df_test.price_paid.apply(lambda x: x.replace("$", ""))
df_test.price_paid = df_test.price_paid.astype(float)
df_test[df_test["price_paid"] > 6]
```
