```python
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('classic')

df = pd.read_csv("../../03.numpy_panda/pandas/data.csv")
```

```python
from matplotlib import colors


df["gender"].value_counts()

# ou 

df["gender"].value_counts(normalize=True).plot.pie()
```

```python
# convertir les valeurs de la colonne "price_paid" en float
df["price_paid"] = df["price_paid"].apply(lambda x: x.replace("$", ""))
df["price_paid"] = df["price_paid"].astype(float)

df.groupby("country")["price_paid"].sum().plot.pie()
```

```python
df.groupby("country").sum(numeric_only=True).plot.bar( rot = 45 , legend = True)
```

differents types de schema :

Line plot (courbe) ------------- Courbe de données continues ------------------ plt.plot()

Bar chart (barres) ------------- Comparaison de catégories -------------------- plt.bar()

Barh (barres horizontales) ----- Idem bar chart, mais horizontal -------------- plt.barh()

Histogram (histogramme) -------- Distribution de données(groupées par classe) - plt.hist()

Scatter plot (nuage de points) - Relation entre deux variables numériques ----- plt.scatter()

Pie chart (camembert) ---------- Proportions de catégories -------------------- plt.pie()

Box plot (boîte à moustaches) -- Répartition statistique (médiane, quartiles) - plt.boxplot()

Area plot (aire) --------------- Courbe avec remplissage ---------------------- plt.fill_between()

Stacked bar chart -------------- Barres empilées pour proportions/groupes ----- plt.bar(..., bottom=...)

Heatmap (carte de chaleur) ----- Représentation matricielle ------------------- plt.imshow() ou seaborn.heatmap()
